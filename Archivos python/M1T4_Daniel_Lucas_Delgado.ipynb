En este ejercicio realizaremos distintas operaciones con string para ver el comportamiento de las cadenas en Python. Si sumamos 2+2 con dos Integer estos se suman, pero si son Strings pasa algo diferente.

```python
# Guardamos el nombre y el apellido en dos variables distintas

nombre = "Daniel"
apellido = "Lucas"

# Ahora realizaremos las 3 operaciones y lo haremos con tres variables distintas

operacion1 = nombre + apellido
operacion2 = nombre + apellido + "."
operacion3 = nombre * 3

# Pero tenemos que mostrarlas por pantalla con un print para ver esas operaciones y verificar que ocurre
print(operacion1)
print(operacion2)
print(operacion3)
```

### Ejercicio 3 Slicing

Sin ninguna duda un comportamiento muy interesante dentro de los strings, que no tienen por que ser una palabra, puede ser un cuadro de texto o una descripción de un producto en amazon, es que podemos seleccionar desde donde queremos empezar a mostrar y donde terminar, pero si un usuario ha introducido un texto podemos no saber necesariamente el tamaño de esa cadena de caracteres para luego hacer operaciones sobre ella.

En este caso podemos señalar con una caja donde epieza y donde termina [inicio:fin] teniendo en cuenta que empezamos desde el 0 y que podemos cojer el -1 como el último caracter. Aunque cabe resaltar que el fin está excluido de la selección.


```python
# Guardamos el nombre completo en un string, pero lo vamos a querer separar
nombreCompleto = "Daniel Lucas"

# En este caso D a n i e l cuenta con 6 carácteres pero empezamos por el 0 y el último se excluye por lo que tendríamos que seleccionar [0:6]
# siendo el 6 el espacio en blanco
print (nombreCompleto[0:6])
```

### Ejercicio 2 Redondeo

En el siguiente ejercicio se nos pide realizar un redondeo a dos decimales. Para esta tarea podemos guardar el resultado en una variable y luego aplicar round(resultado, nº de decimales), de esta manera imprimiremos el resultado solo con dos decimales.

```python
resultado = (365/12) * 14.7

print(f"El resultado es: {resultado}, sin redondear")

# Podemos establecer un formato especifico para que el resultado siempre salga redondeado a dos decimales
formatoResultado = round (resultado, 2)

print(f"El resultado es: {formatoResultado}, redondeado a dos decimales")
```

### Ejercicio 3 Formateo de números
En esta tarea añadiremos dígitos adicionales a una variable para darle distintas formas.

### Ejercicio 4 Formas de formatear

En este caso nos piden distintas cosas que debemos tener en cuenta. Nos piden que pidamos dos parámetros altura y peso. En este caso tenemos que usar el input y guardar dichas variables para luego mostrarlas. Pero la forma en la que se muestra es de una manera específica. Por ello tendremos que usar los distintos conocimientos que hemos adquirido durantes esta práctica y tema.

```python
altura = float(input("Introduce tu altura: "))
peso = float(input("Introduce tu peso: "))

# Ahora mostraremos los datos formateados de distintas maneras

# Podemos hacerlo con .format señalando en las llaves como queremos mostrar los dígitos y el orden en el paréntesis
print("La altura es de {:.2f} metros y el peso es de {:.3f} KG".format(altura, peso))
# Podemos establecer el orden específico señalando 0 y 1 siendo la altura 0 y peso 1
print("La altura es de {0:.2f} metros y el peso es de {1:.3f} KG".format(altura, peso))
# Podemos utilizar abreviaciones para los datos
print("La altura es de {alt:.2f} metros y el peso es de {pes:.3f} KG".format(alt=altura, pes=peso))

# Por último podemos usar lo que hemos hecho en otros ejercicios el f-string que es más elegante
print(f"La altura es de {altura:.2f} metros y el peso es de {peso:.3f} KG")
```
