## Ejercicio 1. División entre 0.
En este apartado tenemos una división entre 0 lo cual nos da error. No siempre se puede controlar las distintas entradas y puede haber un comportamiento inesperado en el programa en el que el resultado sea 0 para una división que tengamos que hacer. Por ello podemos manejar esta casuística mediante try/except especificando el ZeroDivisionError. Es importante que el error se produzca dentro del bloque try para que se captures después el except.

```python

try:
    numero = 7/0
    print(numero)
except ZeroDivisionError as e:
    print("Error, no se puede dividir por cero:", e)
```

## Ejercicio 2. Índice fuera de rango.
En este caso la lista puede ir fuera del bloque del try/except. El problema viene cuando intentamos acceder a ese índice inexistente, si esto fuera una función que accede a un ítem de una lista deberiamos incluir dentro de la propia función este posible error para prevenir que el programa pare de ejecutarse.

```python
lista = [4, 7, 30, 23, 5]

try:
    lista[10]

except IndexError as e:
    print("Error, indice fuera de rango:", e)
```

## Ejercicio 3. Clave inexistente en diccionario
De la misma manera que ocurre con una lista, nos puede ocurrir con un diccionario. Si aplicamos un filtro en el que queramos buscar una clave dentro del diccionario y esta no existe por el motivo que sea nuestro programa podría quedarse bloqueado, por ello añadimos KeyError para notificar de que dicha clave no exite y que nuestro programa continue.

```python
paises = {"españa": "español", "eeuu": "inglés", "italia": "italiano"}
try:
    paises["alemania"]
except KeyError as e:
    print("Error al acceder a la clave, esta no existe:", e)
```

## Ejercicio 4. Error de tipo.
En este caso nos encontramos con una operación que no puede realizarse porque los valores son de distinto tipo. Esto nos lanza un TypeError que podemos subsanar gracias al bloque try/except informando de que es lo que esta ocurriendo y evitando el cierre de forma abrupta del programa.

```python
try:
    resultado = "2" + 10
except TypeError as e:
    print("Error de tipado, no se puede sumar una frase a un número entero:", e)
```

## Ejercicio 5. Claves de entrada de usuario
En este ejercicio he decidido añadir un poco más de complejidad para practicar el manejo de errores. Primero declaramos la variable datos, que contiene el diccionario proporcionado en el enunciado. Después implementamos un bucle while que se ejecutará hasta que el usuario introduzca una clave válida.

Dentro del bucle utilizamos un bloque try/except. En el bloque try se solicita al usuario una clave y se intenta acceder al valor correspondiente dentro del diccionario. Si la clave existe, se imprime su valor y el programa finaliza correctamente.

Si la clave no existe, se produce un KeyError, que es capturado por el bloque except. En ese caso se muestra un mensaje de error indicando que la clave no es válida y se muestran las claves disponibles para que el usuario pueda volver a intentarlo. Gracias a este control de errores, evitamos que el programa se detenga de forma abrupta y garantizamos una interacción más clara y segura.

```python
# Declaración de variable Datos
datos = {
  "nombre": "Juan",
  "edad": 30,
  "ciudad": "Madrid"
}

# Bucle While que se ejecutará hasta que la clave sea válida
while True:
    # Manejo de errores try/except en el que se pide una clave y si no es válida se notifica
    try:
        clave = input("Introduce una clave para obtener su valor: ")
        print(f"El valor asociado a '{clave}' es: {datos[clave]}")
        break
    
    except KeyError as clave_invalida:
        print("Error, la clave introducida no existe en el diccionario:", clave_invalida)
        print("Las claves válidas son:")
        # Bucle para recorrer el diccionario y mostrar las claves válidas
        for clave_valida in datos:
            print(clave_valida)
```

## Ejercicio 6. Conversión y división.
En esta función controlamos los posibles errores que pueden producirse durante el cálculo. Por un lado, el usuario podría introducir letras o valores no numéricos, lo que provocaría un ValueError al intentar convertirlos con int(). Por otro lado, si el segundo número es 0, se produciría un ZeroDivisionError al intentar realizar la división.

Para evitar que el programa se detenga, envolvemos el cálculo dentro de un bloque try/except. Si todo va bien, guardamos el resultado en una variable y lo devolvemos. Si ocurre alguno de los errores mencionados, el bloque except los captura y muestra un mensaje informando del problema. El resto del código puede mantenerse igual, ya que la función se encarga de gestionar los fallos. Personalmente prefiero manejar los errores dentro de la función para tener una mayor claridad sobre el código.


```python
def calcular(numero1, numero2):
    try:
        resultado = (int(numero1) ** 2) / int(numero2)
        return resultado
    except (ZeroDivisionError, ValueError) as e:
        print("No se puede calcular debido al error:", e)
        return None



n1 = input("Introduce el primer número: ")
n2 = input("Introduce el segundo número: ")
resultado = calcular(n1, n2)
print(f"El resultado del cálculo es: {resultado}")
```

## Ejercicio 7. Función para agregar sin repetir
En este ejercicio la función se encarga de añadir un elemento a la lista siempre que no esté ya incluido. Para ello comprobamos si la palabra introducida se encuentra en lista_prueba. Si ya existe, lanzamos un ValueError de forma manual, ya que Python permite añadir elementos duplicados y debemos controlar este caso nosotros mismos. Si la palabra no está en la lista, simplemente la añadimos con append().

```python
def agregar_sin_repetidos(palabra, lista_prueba):
    # Realiza una comprobación para saber si el elemento está ya dentro de la lista, en caso contrario lo añade
    if palabra in lista_prueba:
        raise ValueError (f"Error: imposible añadir elementos duplicados => {palabra}")

    else:
        lista_prueba.append(palabra)
        print("Elemento agregado a la lista:", palabra)


def main():
    lista_prueba = ["Lucario", "Umbreon", "Vivillon"]
    palabra = input("Introduce un elemento en la lista:")
    agregar_sin_repetidos(palabra, lista_prueba)
    print(lista_prueba)

if __name__ == "__main__":
    main()
```

## Ejercicio 8. Investigación de excepciones
En este ejercicio he elegido provocar un FileNotFoundError, ya que todavía no hemos trabajado con programación avanzada pero sí podemos reproducir este tipo de error fácilmente. Para ello, he creado una función que intenta abrir un archivo que no existe. Al hacerlo, Python lanza automáticamente un FileNotFoundError, que capturamos con un bloque try-except para mostrar el mensaje por defecto y un mensaje personalizado. Este error es muy habitual cuando trabajamos con archivos externos, ya que sus rutas pueden cambiar o el archivo puede haber sido eliminado o movido.

```python
def funcion_erronea():
    try:
        open("C:/archivo_que_no_existe.txt")
    except FileNotFoundError as e:
        print("Error, el archivo no existe o la ruta es errónea:", e)

def main():
    print("Vamos a intentar abrir un archivo que no existe y que va a lanzar el error: FileNotFoundError.")
    funcion_erronea()

if __name__ == "__main__":
    main()
```
