## Ejercicio 1. Clase Alumno

```python
class Alumno:
    def __init__(self, nombre, nota):
        self._nombre = nombre
        self._nota = nota
        print("Alumno creado correctamente.")
    
    # Setter y Getter Nombre
    @property
    def nombre(self):
        return self._nombre

    @nombre.setter
    def nombre(self, value):
        self._nombre = value

    # Setter y Getter de Nota
    @property
    def nota(self):
        return self._nota
    
    @nota.setter
    def nota(self, value):
        self._nota = value

    def calificacion(self):
        if self.nota < 5 and self.nota >= 0:
            return "Suspenso"
        elif self.nota >= 5 and self.nota <= 10:
            return "Aprobado"
        else:
            return "Nota no válida"

    def __str__(self):
        return f"Alumn@ {self.nombre} ha sacado un {self.nota}"



# EXPERIMENTACIÓN
# Solo modificar mediante getter y setter
alumno1 = Alumno("Daniel", 4.99)
alumno1.nombre = "Rodrigo"
alumno2 = Alumno("Sara", 10)
alumno3 = Alumno("Cristian", 11)
print(alumno1)
print(alumno1.calificacion())
print(alumno2)
print(alumno2.calificacion())
print(alumno3)
print(alumno3.calificacion())
```

En este código hemos creado la clase alumno y hemos establecido el constructor con atributos privados. Esto se hace mediante el guión bajo delante de cada variable.
De esta manera podemos usar el decorador `@property` para crear los getters y  con `@atributo.setter` los setters. De esta manera accedemos desde fuera con la variable nombre y no con la variable _nombre, eso implica que el encapsulamiento esta correctamente creado.

Después nos encontramos con el método `calificacion()`, el cual se va a encargar de notificar si el alumno en cuestión está aprobado o no. Añadimos también un caso en el que la nota no es válida si es menor que 0 o mayor que 10. Para ver la calificación por pantalla es necesario usar `print()`, porque el método solo devuelve un valor, no lo muestra.

Para mostrar la información del alumno y acceder al objeto creado hemos hecho uso del decorador `__str__`, pero dentro del propio código, si añadimos en cualquier `return self`, podemos acceder al objeto al imprimir nuestro mensaje. Por ejemplo, si cuando imprimimos por pantalla "Alumno creado correctamente" añadimos un `, self`, se mostrará por pantalla el alumno y su nota. Este comportamiento me ha parecido curioso y digno a tener en cuenta.

Por otro lado hemos creado los objetos y hemos añadido las pruebas pertinentes para modificar el nombre mediante la variable que no es privada, todo ello al final del código, dando por finalizada correctamente la tarea.

## Ejercicio 2. Clase Producto

```python
class Producto:
    # Define un constructor con un print
    def __init__(self, codigo, nombre, precio, tipo):
        self._codigo = codigo
        self._nombre = nombre
        self._precio = precio
        self._tipo = tipo
        print("Producto creado correctamente.")

    #Experimentacion: establece getter y setter para probar a añadir objetos
    @property
    def codigo(self):
        return self._codigo
    @codigo.setter
    def codigo(self, value):
        self._codigo = value
    
    @property
    def nombre(self):
        return self._nombre
    @nombre.setter
    def nombre(self, value):
        self._nombre = value
    
    @property
    def precio(self):
        return self._precio
    @precio.setter
    def precio(self, value):
        self._precio = value
    
    @property
    def tipo(self):
        return self._tipo
    @tipo.setter
    def tipo(self, value):
        self._tipo = value

    # Metodo calcular
    def calcular_total(self, cantidad):
        total = cantidad * self.precio
        return f"{self.nombre} x {cantidad} = {total}€"
    
    # Estructura __str__ para que muestre Producto
    def __str__(self):
        return f"-----------\nCódigo: {self.codigo}\nNombre: {self.nombre}\nPrecio: {self.precio}€\nTipo: {self.tipo}\n"



producto1 = Producto("001", "Toallitas", 2, "Higiene")
producto2 = Producto("002", "Pañales", 10, "Higiene")
producto1.nombre = "Pasta de dientes"
print(producto1)
print(producto2)
print(producto2.calcular_total(5))
```

Como en el anterior ejercicio, establecemos todos los atributos del objeto en una clase `__init__()` constructor y aplicamos la privatización de cada uno de ellos añadiéndo el guión bajo. Cuando se crea el objeto la consola muestra "Producto creado correctamente". Después creamos un getter y un setter para cada uno de los atributos y comprobamos que podemos cambiar el nombre del producto "Toallitas" por "Pasta de dientes" gracias al setter, que funciona correctamente.

A continuación con el método `__str__()` damos formato al mensaje que imprimira por pantalla cada uno de los objetos. Para realizarlo usaremos \n y guiones para separar cada producto y cada uno de sus atributos, haciendo uso del f-string y los propios getters.

Finalmente creamos el método calcular_total al cual se le va a pasar el objeto, el precio en se de cada objeto y lo va a multiplicar por una cantidad x que devolverá el valor total del producto en cuestión. También le hemos dado algo de formato para que no arroje únicamente 50 en nuestro caso.

## Ejercicio 3. Clase Empleado

```python
class Empleado:
    def __init__(self, id, nombre, puesto, salario):

        self._id = id
        self._nombre = nombre
        self._puesto = puesto
        self._salario = salario
        print("Empleado creado correctamente")

    @property
    def id(self):
        return self._id
    @id.setter
    def id(self, value):
        self._id = value

    @property
    def nombre(self):
        return self._nombre
    @nombre.setter
    def nombre(self, value):
        self._nombre = value

    @property
    def puesto(self):
        return self._puesto
    @puesto.setter
    def puesto(self, value):
        self._puesto = value

    @property
    def salario(self):
        return self._salario
    @salario.setter
    def salario(self, value):
        self._salario = value

    def incrementar_salario(self, porcentaje):
        salario_incrementado = self._salario + (self.salario * porcentaje *0.01)
        self.salario = int(salario_incrementado)
        return f"Salario incrementado en un {porcentaje}%: {int(salario_incrementado)}€" 

    def mostrar_informacion(self):
        print(f"-----------\nID: {self.id}\nNombre: {self.nombre}\nPuesto: {self.puesto}\nSalario: {self.salario}€\n")

empleado1 = Empleado("3788", "Andrés", "Director", 50000)
empleado2 = Empleado("3789", "Sara", "Jefa RRHH", 45000)
empleado1.nombre = "Andrés García"
empleado1.mostrar_informacion()
empleado2.mostrar_informacion()
print(empleado1.incrementar_salario(5))
empleado1.mostrar_informacion()
```

Como en los ejercicios anteriores, en este caso el procedimiento es el mismo de cara a crear el constructor, los getters y setters y los mensajes por pantalla.
En cuanto al método especial `__str__` no lo hemos usado pero hemos creado un método `mostrar_informacion()` que hace prácticamente lo mismo solo que hay que especificar que imprima por pantalla lo que queramos cuando lo invoquemos. La información la recibimos del propio objeto llamando a los getters para que nos muestre cada uno de los atributos.
El otro método que necesitamos hacer es el de `incrementar_salario(porcentaje)`, que en este caso le hemos añadido self, para poder acceder al salario del objeto que queremos mirar y modificar el salario antiguo por el nuevo. Para obtener el incremento guardamos en una variable el salario incrementado, sumando el salario actual más el porcentaje multiplicado por 0,01 para que de la cantidad en dinero directamente.
En los prints podemos apreciar que se muestra después la información actualizada al ìncrementar_salario()`.
