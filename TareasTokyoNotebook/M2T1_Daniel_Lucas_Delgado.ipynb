## 2.-Índices de cadena
En este ejercicio vamos a crear un texto e imprimir la primera letra usando su índice y la última de la misma manera usando índice negativo. A modo de repaso en Python comenzamos en el índice 0 pero si no sabemos la longitud de nuestro texto podemos acceder a la última posición mediante números negativos, siendo el -1 la última posición.

```python
texto = "Python"

print(texto[0]) #Imprime primera posición "P"
print(texto[-1]) #Imprime la última posición "n"
```

## 3.- Extraer partes de una dirección (slicing)
En este ejercicio vamos a acceder a distintas partes de un texto usando slicing. En las cadenas podemos indicar hasta que posición queremos acceder de esta misma indicando por donde empezar, por ejemplo 0 y por donde terminar, pero sin incluir este último valor.
Por ejemplo si queremos solo coger los 4 primeros valores de la cadena quedaría [0:4], usando 4 en la parte del fin ya que llega a la posición 3 que es en total 4 elementos de la cadena. También podemos llegar al final de la cadena dejando en blanco el segundo espacio [0:].

```python
direccion = "Calle Falsa, 123, Springfield"

# Al poner extraer vamos a guardar las variables que queremos obtener en sitios distintos antes de imprimirlos
nombreCalle= direccion[0:11] #Calle Falsa

numero = direccion[13:16] # 123

ciudad = direccion[18:] # Springfield

print(nombreCalle)

print(numero)

print(ciudad)
```

## 4.-Identificador único con slicing y concatenación
En este ejercicio vamos a establecer un identificador con los tres primeros carácteres de las tres variables del usuario: nombre, apellido y ano_nacimiento.


```python
nombre = "Daniel"

apellido = "Lucas"

ano_nacimiento = "1993"

identificador = nombre[0:3] + apellido[0:3] + ano_nacimiento[-2:]

print(identificador) #DanLuc93
```

## 5.- Inmutabilidad: modificar una cadena
En esta ocasión vamos a tener una palabra, vamos a intentar acceder a un carácter que no estará incluido en nuestra variable. Nos tocará obtener el resultado que queremos con una nueva variable usando concatenación y slicing.

```python
palabra = "robot"

#palabra[0] = "R" # Esto nos arroja: TypeError: 'str' object does not support item assignment
# Las cadenas de texto son inmutables, lo que significa que no se pueden modificar después de haber sido creadas.
# Para que se vea robot usando slicing y concatenación usaremos una nueva variable

palabra_modificada = "R" + palabra[1:] #Robot
print(palabra_modificada)
```

## 6.- Analizador de palabras

Este último ejercicio será un poco más complejo. En primer lugar tenemos dos variables, texto y objetivo. Los tenemos que convertir en minúsculas, vamos a ver cuantas veces aparece una palabra determinada como palabra exacta, con el metodo split, y como subcadena con el metodo count.

```python
texto = "Python es POTENTE. Aprender python es útil. PYTHON abre puertas."
objetivo = "Python"

#primero pasamos las variables a lowerCase para transformar mayúsculas en minúsculas y que sea un factor que no nos afecte
textoMinus = texto.lower()

objetivoMinus = objetivo.lower()

print(len(textoMinus.split(objetivoMinus))) # Hacemos len de ['', ' es potente. aprender ', ' es útil. ', ' abre puertas.']

print(textoMinus.count(objetivoMinus))
```

Podemos observar que no lanzan los mismos resultados. Cuando contamos mediante .split() lo que estamos haciendo realmente es contar los trozos sobrantes del texto, es decir el texto - las veces que aparece la palabra objetivo. Ello hace que aunque corte 3 veces divide la cadena en 4 trozos. Esto se deber al funcnionamiento de slpit, que al cortar la primera vez lo que deja lo guarda en una variable como '', por eso el conteo a través de count es más preciso.

```python
objetivoMayus = objetivo[1:]
textoMod = texto.replace(objetivoMayus, objetivoMayus.upper())

print(textoMod) # PYTHON es POTENTE. Aprender pYTHON es útil. PYTHON abre puertas.
```

Lo que hacemos es sustituir ython por YTHON de tal manera que en el texto original sea reemplazado donde aparece, tanto python Como Python. Condicionales podríamos añadir una segunda transformación para cambiar ese 'pYTHON' por 'PYTHON'.
