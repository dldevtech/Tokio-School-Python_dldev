## Ejercicio 1. Sumatoria de no negativos
En este ejercicio se nos pide crear una función que reciba una lista de números y devuelva la suma de los números no negativos. Para testear esta función hemos creado distintas pruebas:
    - Sumar listas vacías, que deberia dar como resultado 0.
    - Sumar listas con solo números negativos, que también debería dar como resultado 0.
    - Sumar listas con números positivos y negativos, que debería dar como resultado la suma de los números positivos.
    - Sumar listas con números positivos, que debería dar el resultado de la suma de todos esos números.

Además de esto hemos creado un test parametizable repitiendo algunos de estos ejemplos.

### m2t1_ejercicio1.py

```python
def sumatoria_positivos(nums):
    total = 0
    for n in nums:
        if n >= 0:
            total += n
    return total
```

### test_m2t5_ejercicio1.py

```python
import pytest
from m2t5_ejercicio1 import sumatoria_positivos

def test_Sumatoria_listas_vacias():
    assert sumatoria_positivos([]) == 0

def test_sumatoria_negativos():
    assert sumatoria_positivos([-5,-6,-4]) == 0

def test_sumatoria_mixtos():
    assert sumatoria_positivos([-5, 5, 6]) == 11

def test_sumatoria_positivos():
    assert sumatoria_positivos([3, 2, 5]) == 10

@pytest.mark.parametrize(
    "nums, esperado",
    [
        ([], 0),
        ([-5,-6,-4], 0),
        ([-5, 5, 6], 11),
        ([3, 2, 5], 10),
        ([6, 3, 1, -10, 100], 110)
        
    ]
)
def test_sumatoria_parametrizada(nums, esperado):
    assert sumatoria_positivos(nums) is esperado 
```

## Ejercicio 2. Palíndromo básico

En el siguiente ejercicio se nos solicita un testeo sobre si un string es palíndromo o no, esto es, una palabra que se escribe de la misma manera al invertirla.

Para realizar las pruebas, lo haremos mediante testeo parametrizable, ademas de añadir un id a cada una de ellas. Según la documentación dentro de los apuntes, gracias a pytest.param() podemos especificar un "id" o "mark" para cada uno de nuestros test.
Dentro de los 8 casos hemos añadido:
    - Una cadena vacía.
    - Una palabra palíndroma.
    - Una palabra palíndroma con mayúscula, la funcion es case sensitive.
    - Una palabra palíndroma con carácteres mixtos mayúsculas y minúsculas.
    - Una frase palíndroma.
    - Una frase no palíndroma.
    - Un número palíndromo.
    - Un número no palíndromo.

Con ello tenemos 8 pruebas distintas, positivas y negativas, parametrizadas, con su correspondiente título que nos aparece al ejecutar pytest.

### m2t5_ejercicio2.py

```python
def es_palindromo(s: str) -> bool:
    s = s.lower().replace(' ', '')
    return s == s[::-1]
```

### test_m2t5_ejercicio2.py

```python
import pytest
from m2t5_ejercicio2 import es_palindromo

@pytest.mark.parametrize(
    "p, esperado",
    [
        pytest.param("", True, id="cadena_vacia"),
        pytest.param("radar", True, id="palabra_minus_palindroma"),
        pytest.param("Radar", True, id="palabra_mayus_palindroma"),
        pytest.param("SomOs", True, id="palabra_mixta_palindroma"),
        pytest.param("Amor a Roma", True, id="frase_palíndroma"),
        pytest.param("No soy palíndromo", False, id="frase_no_palíndroma"),
        pytest.param("33", True, id="numero_palindromo"),
        pytest.param("34", False, id="numero_no_palindromo"),
        
    ],
)
def test_es_palindromo(p, esperado):
    assert es_palindromo(p) is esperado
```

## Ejercicio 3. Máximo control de errores.
La función maximo_seguro(nums) recibe una lista de números (enteros o floats) y devuelve el valor máximo.

Primero comprueba si la lista está vacía. En ese caso, lanza un ValueError, ya que no es posible obtener un máximo de una lista sin elementos. Si la lista contiene valores, inicializa la variable maximo con el primer elemento.

Después recorre el resto de la lista y actualiza maximo cada vez que encuentra un número mayor. Finalmente devuelve el valor máximo encontrado.

Para validar el comportamiento de la función hemos creado tres pruebas unitarias:

- Caso con números enteros, comprobando que devuelve el máximo correcto.

- Caso con números float, verificando que funciona igual con valores decimales.

- Caso de lista vacía, donde usamos pytest.raises(ValueError) para asegurar que la función lanza la excepción adecuada cuando no hay elementos.

### m2t5_ejercicio3.py

```python
def maximo_seguro(nums):
    if not nums:
        raise ValueError('lista vacía')
    maximo = nums[0]
    for n in nums[1:]:
        if n > maximo:
            maximo = n
    return maximo
```

### test_m2t5_ejercicio3.py

```python
from m2t5_ejercicio3 import maximo_seguro
import pytest

def test_numeros_enteros():
    assert maximo_seguro([3,5,2,10]) == 10

def test_numeros_float():
    assert maximo_seguro([2.0,5.0,100.0]) == 100.0

def test_lista_vacia():
    with pytest.raises(ValueError):
        maximo_seguro([])
```

## Ejercicio 4. Contador simple de palabras.
En este ejercicio, la función recibe una frase o texto y construye un diccionario donde se almacenan las palabras encontradas.
Cada palabra se normaliza a minúsculas y se cuenta cuántas veces aparece en el texto.
Finalmente, la función devuelve un diccionario con cada palabra y su número de repeticiones.

Las pruebas realizadas son las siguientes:

Texto vacío: debe devolver un diccionario vacío.

Palabras repetidas: "Eco eco eco" debe devolver {"eco": 3}.

Caso con mayúsculas y minúsculas: "Hola hola" debe devolver {"hola": 2}, ya que la función normaliza el texto.

Estos son los tests que se nos piden.
Los he implementado mediante test parametrizable y asignando un id a cada caso, para repasar las distintas opciones que tenemos al a hora de testear.

### m2t5_ejercicio4.py

```python
def contar_palabras(texto: str) -> dict:
    conteo = {}
    for palabra in texto.lower().split():
        conteo[palabra] = conteo.get(palabra, 0) + 1
    return conteo
```

### test_m2t5_ejercicio4.py

```python
from m2t5_ejercicio4 import contar_palabras
import pytest

@pytest.mark.parametrize(
    "texto, diccionario",
    [
        pytest.param("", {}, id="cadena_vacia"),
        pytest.param("Eco eco eco", {"eco":3}, id="cadena_palabras_repetidas"),
        pytest.param("Hola hola", {"hola":2}, id="cadena_Hola_hola"), 
    ],
)
def test_contar_palabras(texto, diccionario):
    assert contar_palabras(texto) == diccionario
```

## Ejercicio 5. Filtrar aprobados.
En este ejercicio trabajamos con una función que recibe una lista de tuplas, donde cada tupla contiene el nombre de un alumno y su nota.
La función recorre esa lista y añade a la lista de aprobados únicamente los nombres cuya nota es mayor o igual que 5.
Si la nota es inferior, ese alumno no se incluye en el resultado.

Para comprobar su funcionamiento hemos realizado dos tipos de pruebas:

    - Una prueba unitaria, que verifica que los nombres aprobados se añaden en el orden correcto. Para ello usamos un caso donde todos los alumnos están aprobados, de forma que el orden de entrada debe coincidir con el de salida.

    - Una prueba parametrizada, donde comprobamos varios escenarios:

        - Un alumno aprobado único.

        - Un caso mixto con aprobados y suspendidos.

        - Una lista vacía.

        - Un caso donde todos los alumnos están suspensos.

La función ha pasado correctamente todos los tests realizados.

### m2t5_ejercicio5.py

```python
def filtrar_aprobados(pares):
    resultado = []
    for nombre, nota in pares:
        if nota >= 5:
            resultado.append(nombre)
    return resultado
```

### test_m2t5_ejercicio5.py

```python
from m2t5_ejercicio5 import filtrar_aprobados
import pytest

def test_mantener_orden():
    pares = [("Diego", 6), ("Cesaria", 8), ("Logan", 7)]
    esperado = ["Diego", "Cesaria", "Logan"]
    assert filtrar_aprobados(pares) == esperado

@pytest.mark.parametrize(
    "pares, esperado",
    [
        pytest.param([("Ana", 5)], ["Ana"], id ="caso_aprobado_unico"),
        pytest.param([("Ana", 5),("Luis", 7), ("Jose", 2)], ["Ana", "Luis"], id ="caso_mixto"),
        pytest.param([ ], [], id ="caso_vacio"),
        pytest.param([("Carlos", 3), ("Iván", 4), ("Sergio", 0)], [], id ="caso_suspensos")
    ]
)

def test_filtrar_aprobados(pares, esperado):
    assert filtrar_aprobados(pares) == esperado
```

## Ejercicio 6. Normalizar email.
La función normalizar_email() recibe un correo electrónico en forma de cadena.
Primero comprueba que el email contiene exactamente un carácter @.
Si no es así (por falta o por exceso), lanza un ValueError.
En caso válido, separa el usuario y el dominio y devuelve ambos normalizados en minúsculas.

Para comprobar su funcionamiento hemos realizado tres pruebas:

    - Una prueba parametrizada, donde evaluamos un caso válido y otro con mayúsculas mezcladas, comprobando que el resultado se devuelve correctamente normalizado y separado por el @.

    - Una prueba unitaria para el caso de un correo sin arroba, donde esperamos que se lance un ValueError.

    - Otra prueba unitaria para el caso de un correo con más de un arroba, donde también debe lanzarse un ValueError.

Es importante fijarse en el flujo de la función y tener claro qué tipo de dato introducimos y qué estructura devuelve.

### m2t5_ejercicio6.py

```python
def normalizar_email(email: str):
    if email.count('@') != 1:
        raise ValueError('email inválido')
    usuario, dominio = email.split('@')
    return usuario.lower(), dominio.lower()
```

### test_m2t5_ejercicio6.py

```python
from m2t5_ejercicio6 import normalizar_email
import pytest

@pytest.mark.parametrize(
    "email, esperado",
    [
        pytest.param("miemail@yaju.com", ("miemail","yaju.com"), id="caso_valido"),
        pytest.param("MIEMAIl@guguel.es", ("miemail","guguel.es"), id="caso_norm_mayus"),
    ]
)

def test_normaliz_email_casos_validos(email, esperado):
    assert normalizar_email(email) == esperado

def test_caso_invalido_sin_arroba():
    with pytest.raises(ValueError):
        normalizar_email("miemailyaju.com")

def test_caso_invalido_doble_arroba():
    with pytest.raises(ValueError):
        normalizar_email("miemail@@yaju.com")
```

## Ejercicio 7. Factorial iterativo con control de errores.
En este ejercicio trabajamos con una función que calcula el factorial de un número.
El factorial básicamente consiste en multiplicar todos los números desde 1 hasta n.
Por ejemplo, si n es 5, sería:
1 × 2 × 3 × 4 × 5 = 120.

Hay dos casos especiales:

    -Si n es 0, el resultado es 1 por definición.

    -Si n es negativo, el factorial no existe, así que la función lanza un ValueError.

Para comprobar que la función funciona bien, hemos hecho dos tipos de pruebas:

    -Una prueba parametrizada, donde probamos varios valores válidos:

        -0 debe devolver 1

        -1 debe devolver 1

        -5 debe devolver 120

    -Una prueba unitaria, donde comprobamos que si pasamos un número negativo, la función lanza un ValueError.

La función ha pasado todos los tests correctamente.

### m2t5_ejercicio7.py

```python
def factorial(n: int) -> int:
    if n < 0:
        raise ValueError('n negativo')
    resultado = 1
    for i in range(2, n+1):
        resultado *= i
    return resultado
```

### test_m2t5_ejercicio7.py

```python
from m2t5_ejercicio7 import factorial
import pytest

@pytest.mark.parametrize(
    "n, esperado",
    [
        (0, 1),
        (1, 1),
        (5, 120),
    ]
)
def test_factorial_parametrizado(n, esperado):
    assert factorial(n) == esperado

def test_num_negativo():
    with pytest.raises(ValueError):
        factorial(-1)
```

## Ejercicio 8. Media aritmética.
Para este ejercicio he tenido que usar pytest.approx().
No aparece en los apuntes, pero sí en la documentación oficial de pytest:
[pytest.approx](https://docs.pytest.org/en/stable/reference/reference.html).
Esta utilidad sirve para comparar números flotantes con un pequeño margen de error, ya que a veces los decimales no coinciden exactamente.

Para probar la función media() he hecho una prueba parametrizada con dos casos:

    -Un array sencillo donde la media da un número exacto.

    -Un array donde la media da un número con muchos decimales (por ejemplo 2.666666…), y ahí es donde approx tiene sentido, porque puedo comparar con un valor aproximado sin exigir que todos los decimales coincidan.

Con ello cuando realizamos el assert en el test, en esperado podemos establecer un margen de error de 0.01 por ejemplo para que no tenga en cuenta el resto de flotantes.

Por otro lado hemos hecho la prueba de la lista vacía en la que nos avisa de ValueError.

## m2t5_ejercicio8.py

```python
def media(nums):
    if not nums:
        raise ValueError('lista vacía')
    return sum(nums) / len(nums)
```

## test_m2t5_ejercicio8.py

```python
from m2t5_ejercicio8 import media
import pytest

@pytest.mark.parametrize(
    "nums, esperado",
    [
        ([3,2,4,8], 4.25),
        ([3,3,2], 2.66),
    ]
)

def test_media_aprox(nums, esperado):
    assert media(nums) == pytest.approx(esperado, 0.01)

def test_lista_vacia():
    with pytest.raises(ValueError):
        media([])
```
