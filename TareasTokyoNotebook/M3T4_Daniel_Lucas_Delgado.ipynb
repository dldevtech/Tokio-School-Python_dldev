## Ejercicio 1. Composición básica
En el siguiente ejercicio se nos plantea una situación de composición básica en la que tenemos una clase Pedido que tiene un objeto cliente, que pertenece a la clase cliente.

Para este caso se ha realizado los siguientes objetivos.

1. Crear la clase Cliente con atributos privados y métodos getter y setter para cada uno de ellos.

2. Crear la clase Pedido con atributos privados y métodos getter y setter para cada uno de ellos, incluyendo el atributo cliente que es un objeto de la clase Cliente.

3. Implementación del método `__str__` para ambas clases.

4. Función `calcular_total()` que calcula el precio total del pedido según la cantidad y el precio_unidad.

Algunas decisiones tomadas durante el proceso han sido sobre todo la generación de un código, que se crea a través de la selección de carácteres de las cadenas de los atributos del cliente. O que a la hora de crear un pedido primero hay que crear un cliente.((Aunque no estoy mu seguro de esto, a lo mejor es mas por que siempre es asi en composicion basica))

Finalmente se realizaron las pruebas que se nos han pedido en la parte de experimentación. 
    - Hemos creado dos clientes y dos pedidos.
    - Hemos cambiado datos del pedido para comprobar los getters y setters de Pedido().
    - Hemos desarrollado el método `__str__()` para mostrar la información sobre el objeto creado.
    - Se han modificado los datos del cliente desde el pedido y se han veríficado dichos cambios.

La única cosa que ha quedado colgando es el código que al cambiar de cliente no se actualiza y que el mismo cliente tiene el mismo código para dos pedidos distintos, no obstante era para seguir realizando pruebas de cara a la práctica con cadenas y manipulación de strings.



```python
class Cliente:
    def __init__(self, nombre, direccion, telefono):
        self._nombre = nombre
        self._direccion = direccion
        self._telefono = telefono
        print(f"Cliente creado con éxito {nombre}: {direccion}, {telefono}")

    @property
    def nombre(self):
        return self._nombre
    @nombre.setter
    def nombre(self, nombre):
        self._nombre = nombre

    @property
    def direccion(self):
        return self._direccion
    @direccion.setter
    def direccion(self, direccion):
        self._direccion = direccion

    @property
    def telefono(self):
        return self._telefono
    @telefono.setter
    def telefono(self, telefono):
        self._telefono = telefono

    def __str__(self):
        return f"Datos del cliente\n-----------------\nNombre: {self.nombre}\nDireccion: {self.direccion}\nTeléfono: {self.telefono}\n-------------"
    
class Pedido:
    def __init__(self, producto, cantidad, precio_unitario, cliente):

        self._producto = producto
        self._cantidad = cantidad
        self._precio_unitario = precio_unitario
        self.cliente = cliente
        self._codigo = (self.cliente.nombre[0]+self.cliente.direccion[7]+self.cliente.telefono[5]).upper()
        print(f"Pedido {self._codigo} para Cliente:{self.cliente.nombre}\n")

    @property
    def producto(self):
        return self._producto
    @producto.setter
    def telefono(self, producto):
        self._producto = producto

    @property
    def cantidad(self):
        return self._cantidad
    @cantidad.setter
    def cantidad(self, cantidad):
        self._cantidad = cantidad

    @property
    def precio_unitario(self):
        return self._precio_unitario
    @precio_unitario.setter
    def precio_unitario(self, precio_unitario):
        self._precio_unitario = precio_unitario
    
    def calcular_total(self, cantidad, precio_unitario):
        return cantidad * precio_unitario
    
    def __str__(self):
        total = self.calcular_total(self.cantidad, self.precio_unitario)
        return f"\n---Datos del pedido---\nCliente: {self.cliente.nombre}\nPedido {self._codigo}\nProducto: {self.producto} -> {self.precio_unitario}€\n\nTotal: {total}€\n"+"-"*22


c1 = Cliente("Jose","Calle Alcalá, 13, 2A","+34 622222222")
print(c1)
p1 = Pedido("Fidget Toy", 5,5, c1)
c2 = Cliente("Daniel","Calle de la Piruleta, 6","+34 633333333")
print(c2)
p2 = Pedido("Fidget Toy", 10, 5, c2)
print(p1)
print(p2)
p2 = Pedido("Teclado", 1, 20, c1)
print(p2)
p1.cliente.nombre = "Daniel Lopez"
print(p1)
```

## Ejercicio 2. Composición avanzada
Creación:
    - Amplía el sistema anterior para que un pedido pueda contener varios productos.
    Crea las clases:
        - Cliente (igual que antes).
        - Producto con los atributos codigo, nombre y precio_unitario.
        - Pedido con los atributos codigo, cliente (objeto Cliente), productos (lista de objetos Producto) y cantidades (lista paralela de enteros).

Implementa los métodos:
    - __str__() → imprime un resumen general del pedido.
    - mostrar_pedido() → muestra la información del cliente y el detalle de cada producto con su cantidad.

Experimentación:
    - Crea un cliente, varios productos y un pedido con listas de productos y cantidades.
    - Muestra la información del pedido y luego la detallada.

```python
class Cliente():
    def __init__(self, nombre, direccion, telefono):
        self.nombre = nombre
        self.direccion = direccion
        self.telefono = telefono
        print("Cliente creado con éxito\n")

    #-------METODO STR-------
    def __str__(self):
        return f"DATOS DEL CLIENTE\n-----------------\nNombre: {self.nombre}\nDireccion: {self.direccion}\nTeléfono: {self.telefono}\n-------------\n"
    


class Producto():
    def __init__(self, codigoPR, nombre, precioUnitario):
        self.codigoPR = codigoPR
        self.nombre = nombre
        self.precioUnitario = precioUnitario
        print("Producto creado con éxito\n")


    #-------MÉTODO STR------
    def __str__(self):
        return f"DATOS DEL PRODUCTO\n-----------------\nCódigo: {self.codigoPR}\nNombre: {self.nombre}\nPrecio Unitario: {self.precioUnitario}€\n-------------\n"

class Pedido():
    def __init__(self, codigoPD, cliente, productos, cantidades):
        self.codigoPD = codigoPD
        self.cliente = cliente
        self.productos = productos
        self.cantidades = cantidades
        print("Pedido creado correctamente\n")

    def calcularTotal(self):
        total = 0
        for i in range(len(self.productos)):
            producto = self.productos[i]
            cantidad = self.cantidades[i]
            precioUnitario = producto.precioUnitario
            total += cantidad * precioUnitario
        return total

    # Creación del método mostrarInformacion(), el cual se encargará de mostrar todos los detalles del Pedido
    def mostrarInfo(self):
        print(f"DATOS DEL PEDIDO\nCódigo Pedido: {self.codigoPD}\nCliente: {self.cliente.nombre}\n")
        total = self.calcularTotal()
        for i in range(len(self.productos)):
            producto = self.productos[i]
            cantidad = self.cantidades[i]
            print(f"{producto.nombre} x {cantidad} unidades")
        print(f"Precio total: {total}€")



    #-------MÉTODO STR-------
    def __str__(self):
        # Mostrar resumen del pedido
        total = self.calcularTotal()
        
        return f"--------------\nRESUMEN DEL PEDIDO: {self.codigoPD}\nA nombre de: {self.cliente.nombre}\nTotal: {total}€\n--------------"
    
c1 = Cliente("Daniel", "Calle 123", "+34 666666666")    # Creación del cliente y mostramos
print(c1)
# Creamos los distintos productos 
productoA = Producto("001", "Manzana", 1)
productoB = Producto("002", "Piña", 2)
productoC = Producto("003", "Pizza", 3)
# Lista de la compra de productos
productos = [productoA, productoB, productoC]   # Lista de cantidades
cantidades = [3, 1, 3]

# Creamos un pedido
pedido1 = Pedido("123", c1, productos, cantidades)
print(pedido1)  # Mostramos el resumen
Pedido.mostrarInfo(pedido1) # Método mostrarInfo()

```

En este caso vamos a realizar una composición distinta:
    - Tenemos una clase Cliente, como antes.
    - Una clase Productos, que son cada uno de los elementos que queremos comprar.
    - Clase Pedido que tiene objetos de cliente y listas de productos, cada uno con sus precios cantidades etc...

## Ejercicio 3. Herencia simple en un sistema de viviendas.
Creación:
    - Crea la clase padre Vivienda con los atributos direccion, metros_cuadrados y precio.
    - Crea las clases hijas:
    - Casa → añade el atributo jardin (booleano).
    - Apartamento → añade el atributo piso (número de planta).
    - Sobrescribe el método __str__() en las clases hijas.

Experimentación:
    -Crea una Casa y un Apartamento, imprime su información y modifica atributos para observar los cambios.

```python
class Vivienda():
    def __init__(self, direccion, metrosCuadrados, precio):
            self.direccion = direccion
            self.metrosCuadrados = metrosCuadrados
            self.precio = precio
            print("VIVIENDA. Clase madre creada\n")

    def __str__(self):
          return f"DATOS VIVIENDA\nDirección: {self.direccion}\nMetros Cuadrados: {self.metrosCuadrados}m²\nPrecio: {self.precio}€\n"
    
class Casa(Vivienda):
    def __init__(self, direccion, metrosCuadrados, precio, jardin):
        super().__init__(direccion, metrosCuadrados, precio)    # Pedimos a Padre que se encargue de esos atributos
        self.jardin = jardin    # Booleano. Tiene o no tiene. Solo la hija
        print("CASA. Clase hija creada\n")

    def __str__(self):
        # Creamos dos variables strings 
        base = super().__str__()    # Almacena el método str del padre
        jardin = f"Jardín:"     # Traducción humana para jardín Sí o No

        # Condicional para mensaje dinámico
        # True SI
        if self.jardin == True:
             jardin += "Sí\n"   # Traducimos a humano
             return base + jardin   # Unimos toda la información
        # False NO
        elif self.jardin == False:
             jardin += "No\n"
             return base + jardin
    
class Apartamento(Vivienda):
    def __init__(self, direccion, metrosCuadrados, precio, piso):
        super().__init__(direccion, metrosCuadrados, precio)
        self.piso = piso
        print("APARTAMENTO. Clase hija creada\n")

    def __str__(self):
        # Apartamento es como padre pero con piso.
        base = super().__str__()
        return base + f"Piso: {self.piso}º"

# EXPERIMENTACIÓN ------------------------------        
# Creación Vivienda y str
vivienda1 = Vivienda("Calle", 100, 450000)
print(vivienda1)
# Creación Casa y str
casa1 = Casa("Calle 2", 120, 400000, False)
print(casa1)
# Creación Apartamiento y str
apartamento1 = Apartamento("Calle 2", 120, 400000, 4)
print(apartamento1)

```

Este caso de herencia es mucho mas sencillo que el anterior.
El problema esta vez es crear un sistema en el que tenemos una clase padre Vivienda() con:
    - direccion
    - metrosCuadrados
    - precio

Por otro lado las clases hijas que compartiran los atributos del padre y aportarán un atributo distintos, son dos tipos distintos de Vivienda.
    - Clase Casa que añade el atributo jardin (booleano)
    - Clase Apartamento que añade el atributo piso.

### Clase Vivienda
Lo primero que realizamos es la creación de la clase padre con los atributos que se nos piden. También he decidido añadir el método __str__ ya creado con los datos básicos.

### Clase Casa
Esta clase será hija de la clase Vivienda(), por lo que cabe destacar que la nomenclatura de atributos debe de ser la misma y en el mismo orden que en el padre, para finalmente añadir los distintos atributos específicos de la clase hija.
Sin embargo, dentro del método __str__(), he decidido añadir un condicional que interprete el valor booleano de jardin. De esta manera en el mensaje aparecerá sí o no, en vez de True o False.
En cada una de las iteraciones nos aseguramos de que efectivamente se crean los objetos y se muestran correctamente.

### Clase Apartamento
El resultado es parecido al de la clase Casa() pero mucho más sencillo. Aqui simplemente informamos del piso pero no tenemos que interpretar nada complejo como en la clase anterior.

### Métodos __str__() actualizados.
En cada una de las clases hijas se ha establecido un string base que hereda el contenido del método __str__() padre. Después anexamos la información característica de cada tipo de vivienda.
