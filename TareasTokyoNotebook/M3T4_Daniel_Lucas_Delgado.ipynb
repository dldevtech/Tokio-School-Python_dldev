## Ejercicio 2. Composición avanzada

`````python
class Cliente():
    def __init__(self, nombre, direccion, telefono):
        self.nombre = nombre
        self.direccion = direccion
        self.telefono = telefono
        print("Cliente creado con éxito\n")

    #-------METODO STR-------
    def __str__(self):
        return f"DATOS DEL CLIENTE\n-----------------\nNombre: {self.nombre}\nDireccion: {self.direccion}\nTeléfono: {self.telefono}\n-------------\n"
    


class Producto():
    def __init__(self, codigoPR, nombre, precioUnitario):
        self.codigoPR = codigoPR
        self.nombre = nombre
        self.precioUnitario = precioUnitario
        print("Producto creado con éxito\n")


    #-------MÉTODO STR------
    def __str__(self):
        return f"DATOS DEL PRODUCTO\n-----------------\nCódigo: {self.codigoPR}\nNombre: {self.nombre}\nPrecio Unitario: {self.precioUnitario}€\n-------------\n"

class Pedido():
    def __init__(self, codigoPD, cliente, productos, cantidades):
        self.codigoPD = codigoPD
        self.cliente = cliente
        self.productos = productos
        self.cantidades = cantidades
        print("Pedido creado correctamente\n")

    def calcularTotal(self):
        total = 0
        for i in range(len(self.productos)):
            producto = self.productos[i]
            cantidad = self.cantidades[i]
            precioUnitario = producto.precioUnitario
            total += cantidad * precioUnitario
        return total

    # Creación del método mostrarInformacion(), el cual se encargará de mostrar todos los detalles del Pedido
    def mostrarInfo(self):
        print(f"DATOS DEL PEDIDO\nCódigo Pedido: {self.codigoPD}\nCliente: {self.cliente.nombre}\n")
        total = self.calcularTotal()
        for i in range(len(self.productos)):
            producto = self.productos[i]
            cantidad = self.cantidades[i]
            print(f"{producto.nombre} x {cantidad} unidades")
        print(f"Precio total: {total}€")



    #-------MÉTODO STR-------
    def __str__(self):
        # Mostrar resumen del pedido
        total = self.calcularTotal()
        
        return f"--------------\nRESUMEN DEL PEDIDO: {self.codigoPD}\nA nombre de: {self.cliente.nombre}\nTotal: {total}€\n--------------"
    
c1 = Cliente("Daniel", "Calle 123", "+34 666666666")    # Creación del cliente y mostramos
print(c1)
# Creamos los distintos productos 
productoA = Producto("001", "Manzana", 1)
productoB = Producto("002", "Piña", 2)
productoC = Producto("003", "Pizza", 3)
# Lista de la compra de productos
productos = [productoA, productoB, productoC]   # Lista de cantidades
cantidades = [3, 1, 3]

# Creamos un pedido
pedido1 = Pedido("123", c1, productos, cantidades)
print(pedido1)  # Mostramos el resumen
Pedido.mostrarInfo(pedido1) # Método mostrarInfo()

`````

En este caso vamos a realizar una composición distinta:
    - Tenemos una clase Cliente, como antes.
    - Una clase Productos, que son cada uno de los elementos que queremos comprar.
    - Clase Pedido que tiene objetos de cliente y listas de productos, cada uno con sus precios cantidades etc...

## Ejercicio 3. Herencia simple en un sistema de viviendas.

`````python
class Vivienda():
    def __init__(self, direccion, metrosCuadrados, precio):
            self.direccion = direccion
            self.metrosCuadrados = metrosCuadrados
            self.precio = precio
            print("VIVIENDA. Clase madre creada\n")

    def __str__(self):
          return f"DATOS VIVIENDA\nDirección: {self.direccion}\nMetros Cuadrados: {self.metrosCuadrados}m²\nPrecio: {self.precio}€\n"
    
class Casa(Vivienda):
    def __init__(self, direccion, metrosCuadrados, precio, jardin):
        super().__init__(direccion, metrosCuadrados, precio)    # Pedimos a Padre que se encargue de esos atributos
        self.jardin = jardin    # Booleano. Tiene o no tiene. Solo la hija
        print("CASA. Clase hija creada\n")

    def __str__(self):
        # Creamos dos variables strings 
        base = super().__str__()    # Almacena el método str del padre
        jardin = f"Jardín:"     # Traducción humana para jardín Sí o No

        # Condicional para mensaje dinámico
        # True SI
        if self.jardin == True:
             jardin += "Sí\n"   # Traducimos a humano
             return base + jardin   # Unimos toda la información
        # False NO
        elif self.jardin == False:
             jardin += "No\n"
             return base + jardin
    
class Apartamento(Vivienda):
    def __init__(self, direccion, metrosCuadrados, precio, piso):
        super().__init__(direccion, metrosCuadrados, precio)
        self.piso = piso
        print("APARTAMENTO. Clase hija creada\n")

    def __str__(self):
        # Apartamento es como padre pero con piso.
        base = super().__str__()
        return base + f"Piso: {self.piso}º"

# EXPERIMENTACIÓN ------------------------------        
# Creación Vivienda y str
vivienda1 = Vivienda("Calle", 100, 450000)
print(vivienda1)
# Creación Casa y str
casa1 = Casa("Calle 2", 120, 400000, False)
print(casa1)
# Creación Apartamiento y str
apartamento1 = Apartamento("Calle 2", 120, 400000, 4)
print(apartamento1)

`````

## Ejercicio 6. Herencia múltiple y polimorfismo en un sistema de vehículos.

`````python
# CLASES PADRE
class Vehiculo ():
    def repostar(self):
        return "Vehículo repostado\n"
        
class Electrico():
    def repostar(self):
        return "Batería cargada con éxito\n"

# CLASES HIJAS que heredan en distinto orden    
class BicicletaElectrica(Electrico, Vehiculo):
    def repostar(self):
        return super().repostar()
    
class QuadHibrido (Vehiculo, Electrico):
    def repostar(self):
        return super().repostar()


# MRO FUNCIÓN-----------------------------
def MRO(hija,self):
    print("MRO:", end=" ")
    for cls in hija.mro():
        print(cls.__name__, end=" ")
    print("\n--- Orden de MRO ---")
    print (self.repostar())

# CREACIÓN DE OBJETOS---------------------
bici = BicicletaElectrica()
quad = QuadHibrido()

# MRO + EXPERIMENTACIÓN
# BicicletaElectrica
MRO(BicicletaElectrica,bici)
# QuadHibrido
MRO(QuadHibrido, quad)
`````

### Clases Padre
- En este caso interpreto por el ejercicio que lo que hacemos es crear dos clases Padre distintas que van a tener un método con el mismo nombre, repostar().
- Al crear las clases hijas BicicletaElectrica y QuadHibrido, establecemos el orden de herencia de distinta forma.

### Clases Hija
- En este caso para heredar repostar lo que hacemos es una HERENCIA COOPERATIVA con super(). El MRO muestra el orden de resolución de métodos, y super() utiliza ese orden para decidir qué método repostar() ejecutar, de esta manera el flujo se vuelve más predecible y mantenible.
- Para comprobar esto, en la parte de experimentación podemos imprimir el MRO de cada clase hija para comprobar el flujo de programación.

# Ejercicio 8. Herencia multinivel y polimorfismo (tema libre).

`````python
class Pan ():
    def __init__(self):
        self.ingredientes = ["Pan"]
        print("Bocadillo creado con éxito\n")
    def __str__(self):
        return f"NIVEL 0. Ingredientes del bocadillo: " + ", ".join(self.ingredientes)

class PanConJamon (Pan):
    def __init__(self):
        super().__init__()
        self.ingredientes.append("Jamon")
    def __str__(self):
        return f"NIVEL 1. Ingredientes del bocadillo de Jamón: " + ", ".join(self.ingredientes)

class PanConJamonQueso (PanConJamon):
    def __init__(self):
        super().__init__()
        self.ingredientes.append("Queso")
    def __str__(self):
        return f"NIVEL 2. Ingredientes del bocadillo de Jamón y Queso: " + ", ".join(self.ingredientes)


class PanConJamonQuesoAceite (PanConJamonQueso):
    def __init__(self):
        super().__init__()
        self.ingredientes.append("Aceite")
    def __str__(self):
        return f"NIVEL 3. Ingredientes del bocadillo Semicompleto: " + ", ".join(self.ingredientes)

class PanConJamonQuesoAceiteyPatatas(PanConJamonQuesoAceite):
    def __init__(self):
        super().__init__()
        self.ingredientes.append("Patatas")
    def __str__(self):
        return f"NIVEL 4. Ingredientes del bocadillo Completo: " + ", ".join(self.ingredientes)
# EXPERIMENTACION ----------------------

# Creación de objetos
bocadillo1 = Pan()
bocadillo2 = PanConJamon()
bocadillo3 = PanConJamonQueso()
bocadillo4 = PanConJamonQuesoAceite()
bocadillo5 = PanConJamonQuesoAceiteyPatatas()

# Demostramos polimorfismo y herencia multinivel
# Se llama al método __str__ pero cada uno imprime algo distinto
print(bocadillo1)
print(bocadillo2)
print(bocadillo3)
print(bocadillo4)
print(bocadillo5)
`````
