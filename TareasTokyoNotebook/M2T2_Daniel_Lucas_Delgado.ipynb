## Ejercicio 1
Necesitamos leer tres números por teclado, evaluar si estan en orden ascendente o no y si el primer elemento es 0 tiene que saltar un error. Nos piden no usar listas como condición.

```python
n1 = int(input("Introduce el primer numero:"))  # Pedimos el primer numero y lo guardamos en una variable n1

while n1 == 0:  # Tenemos que verificar si el primer número es 0
    print("El primer número no puede ser 0")    # si no notificar del error
    n1 = int(input("Vuelve a introducir el primer número:"))    # volver a pedir

# Pedimos los otros dos números
n2 = int(input("Introduce el segundo numero:"))
n3 = int(input("Introduce el tercer numero:"))

# Verificamos si están en orden ascendente 
if n1<n2<n3:
    print("Los numeros estan de forma ascendente")

# En caso contrario pedir que pruebe de nuevo
else:
    print("Prueba otra vez de forma ascendente")

print(f"Los números introducidos son: {n1}, {n2}, {n3}.")
```

## Ejercicio 3. Sumatorio de los primeros N números

En este ejercicio se nos pide que, dado un número introducido por el usuario, generemos un sumatorio desde 0 hasta ese número sin incluirlo. Para ello usamos un bucle while con un contador que va añadiendo cada valor a una lista, y al final sumamos todos los elementos.
La función sum nos permite obtener la suma total de los valores almacenados en la lista y mostrar el resultado final.

```python
n = int(input("Introduce un numero: ")) # Pedimos el numero al usuario, que va a ser n
sumatorio = []  # Creamos la lista donde se van a guardar los números que ya de base están en int
i = 0   # Creamos una variable contador o indice que va a aumentar hasta llegar al número n

while i < n:   # Y como no sabemos que numero va a introducir el usuario usamos while hasta que i valga lo mismo que n
    sumatorio.append(i) # Añadimos a la lista cada número de cada iteración
    i+=1     # Aumentamos el contador en uno para que se repita

print(sumatorio)    # Esta sería la lista después de las iteraciones
print(sum(sumatorio))   # Con sum podemos hacer una suma de todo lo que tenemos dentro de la lista
```

## Ejercicio 4. Contador de letras "a".
Necesitamos saber cuantas veces introduce la letra a el usuario, y se introducen una por una. Termina cuando introducimos punto y al finalizar introduce el número total de veces que se ha introducido la letra a.
Para ello vamos a declarar las variables primero, después mostramos por pantalla la información sobre el programa.
A continuación, abrimos un bucle while que se repetira hasta que añadimos un '.'. Pedimos la letra por teclado y pedimos dos condiciones, que si el usuario mete más de una letra se le avise y que si el usuario introduce a o A se sume uno al contador que se mostrará al final del programa. En este caso lo he hecho sensible a mayúsculas, pero si quitamos esa condición el programa solo sumaria las 'a' minúsculas.

```python
# Primero declaramos la letra y el contador
letra = ""
contadorLetras = 0

# Mostramos por consola cómo funciona el programa
print("Voy a contar cuantas veces escribes la letra 'a', para salir escribe'.'")

#Abrimos el While que se repetirá hasta que el usuario ingrese .
while letra != ".":
    letra = input("Escribe una letra: ")

    # Hacemos una condición que avise al usuario si mete mas de una letra
    if len(letra) > 1:
        print("Introduce solo una letra")
        continue

    # Hacemos otra condición para sumar 1 si la letra añadida es a o A.
    if letra.lower() == "a":
        contadorLetras +=1  # Sumamos 1 al contador
    
# Mostramos por pantalla el resultado final    
print(f"Has introducido la letra a {contadorLetras} veces")
```

## Ejercicio 5. Contador de letras "a" en palabras.

En este ejercicio se nos pide que, dada una palabra introducida por el usuario contemos cuántas letras a aparecen y que mostremos un contador total de las apariciones de la letra hasta que el usuario escriba un '.'. Para ello usamos un bucle while que solicita palabras de forma repetida y finaliza cuando se introduce un '.'.
Para contar el número de letras 'a' recorremos cada carácter con un bucle for, normalizando mayúsculas y tildes para evitar diferencias. En cada iteración informamos al usuario de que es lo que ha introducido y como va su contador.

```python
# Primero declaramos la variable palabra vacía y el contador desde 0
palabra = ""
contador = 0

# Después abrimos un bucle del que saldremos cuando se introduzca "."
while palabra != ".":
    # Solicitamos la palabra al usuario
    palabra = input ("Introduce una palabra, te diré cuantas letras 'a' aparecen: \n -> Para salir del programa escribe .")

    # Añadimos la condición de que si escribe "." se cierra el programa para que no se imprima la palabra introducida es "."
    if palabra == ".":
        print("Gracias por utilizar el servicio.")
        break
    
    # Recorremos los caracteres de la palabra y lo formateamos a minúscula y eliminamos las tildes si las hubiera
    for letra in palabra:
        if letra.lower().replace("á", "a") == "a":
            contador += 1   # Aumentamos el contador
    
    #Informamos al usuario de lo que ha escrito y como va el contador de letras
    print (f"Has acumulado {contador} 'a'")
    print(f"Tu palabra era: {palabra}")
```

## Ejercicio 6. Sumatorio de números pares decrecientes.
El ejercicio dice que sumemos los números pares comprendidos entre 100 y 73 en orden descendente. Tenemos que hacerlo con for o while.

Lo primero que hacemos es declarar la variable donde vamos a hacer el sumatorio, después en los propios apuntes de M2T2 podemos observar como hay un ejemplo de una cuenta regresiva de 10 a 0 usando el bucle for con in range. 

Primero ponemos donde empieza, después donde acaba y por último el rango que se puede establecer. Por eso empezamos en 1000, acabamos en 72, porque acaba en 73 y el rango es -2 para que vaya de dos en dos estableciendo solo los numeros pares pero de forma negativa porque va en orden descendente.

```python
print("Voy a calcular la suma total de los números pares que hay desde 1000 hasta 73")
# Declaramos variable donde vamos a sumar
sumaPares = 0
# Con el for establecemos un rango indicando inicio, fin y cada cuanto
for i in range(1000,72,-2):
    # Sumamos al total i en cada iteración
    sumaPares += i
# Imprimimos el resultado por pantalla
print(sumaPares)
```

## Ejercicio 7. Lista de números: análisis y operaciones.
Se nos pide que solicitemos al usuarios cuántos valores va a introducir, por ello entiendo que es libre y vamos a pedir en este caso 5 valores. Después vamos a guardar esos valores en una lista y tenemos que ordenarlor de menos a mayor, calcular el promedio e identificar el valor máximo y mínimo, además de indicar la posición en la que se encuentra.

En este ejercicio el programa solicita al usuario diez números y los almacena en una lista. A medida que se introducen, se va acumulando su suma para poder calcular el promedio posteriormente. En el ejercicio 3 hemos usado la función sum, pero aqui lo hemos hecho manualmente. Una vez recogidos todos los valores, se muestra la lista original y también una versión ordenada de menor a mayor, sin modificar la lista inicial para conservar las posiciones originales de cada número.

Después se calcula el promedio dividiendo el sumatorio total entre la cantidad de elementos de la lista. Finalmente, se identifica el valor mínimo y el máximo usando las funciones min() y max(), y se obtiene la posición en la que fueron introducidos mediante el método index(), ajustando el índice para mostrarlo en formato humano (empezando en 1 en lugar de 0).



```python
# Declaramos las variables que necesitamos, la lista de numeros y el sumatorio de estos para calcular el promedio
listaNumeros = []
sumatorioLista = 0

print("Introduce 10 números")

# Recorremos un bucle 10 veces que es el numero de veces que vamos a pedir los números al usuario
for i in range(10):
    numero = int(input(f"Introduce el número {i+1}: "))
    listaNumeros.append(numero)
    sumatorioLista += numero
print(listaNumeros)

# Gurdamos la lista ordenada en una variable nueva para mantener las posiciones iniciales de los números
listaOrdenada = sorted(listaNumeros)
print(f"La lista ordenada es {listaOrdenada}")

# Calculamos el promedio de la lista que en este caso la media va a ser siempre de 10 números pero aun asi lo haremos con len
promedio = sumatorioLista/len(listaNumeros)
print(f"El promedio de esta lista es: {promedio}")
# Por último señalamos el número mínimo y el número máximo con min y max de la lista original, de esta manera podemos acceder al
# índice para señalar en que posición se ha introducido, traducido a las posiciones humanas.
print(f"El número mínimo es {min(listaNumeros)} y se ha introducido en la posición {listaNumeros.index(min(listaNumeros))+1}")
print(f"El número máximo es {max(listaNumeros)} y se ha introducido en la posición {listaNumeros.index(max(listaNumeros))+1}")
```

## Ejercicio 8. Menú compra-venta

Vamos a crear primero las distintas variables que nos piden: en este caso, tres listas diferentes, una para la compra, otra para la venta y otra para el alquiler. Después mostramos el menú con las opciones. En un principio lo mostraba en cada iteración, pero eran demasiadas salidas por consola, así que he decidido mostrarlo solo al principio y también cuando ninguna de las opciones coincide, por si el usuario introduce algo que no toca (como una frase o dos números seguidos).

Durante el testeo de la aplicación me di cuenta de que, si no se implementaba un manejo de errores, la aplicación seguía en ejecución sin pedir nuevamente opciones. Por eso añadí case _: como manejo de error, mostrando el menú una vez más.

Después de eso, para que el programa se repitiera, usamos un bucle while que continúa ejecutándose hasta que el usuario introduce “5”, en cuyo caso deja de pedir opciones. Valoré usar un while True, pero con esa opción me encontraba con que el programa seguía ejecutándose constantemente a pesar del break, así que añadí la condición de que se repitiera siempre que opcion fuera distinta de "5".

Lo único que queda por hacer es construir el menú con un match-case: dependiendo de la opción introducida, se añaden elementos a cada una de las listas o se muestran estas últimas. También es necesario volver a pedir al usuario una nueva opción para que el programa siga repitiéndose.

Al introducir el “5”, nos despedimos y cerramos la ejecución con el break. Como opcion es "5", se cierra todo el bucle.


```python
# Primero vamos a estableces todas las listas vacías que necesitamos y mostraremos por pantalla el menu
listaCompra = []
listaVenta = []
listaAlquiler = []

print("MENÚ COMPRA-VENTA\n1. Comprar\n2. Vender\n3. Alquilar\n4. Mostrar listas de datos\n5. Salir")
opcion = input("Introduce tu opción:")

# Abrimos el bucle siempre que la opción sea distinta a 5
while opcion != "5":
    match opcion:
    
        # Caso de que quiera comprar algo, se añade a la lista y se pide opción nueva
        case "1":
            item = input(("Introduce que elemento quieres comprar: "))
            listaCompra.append(item)
            opcion = input("Introduce tu opción:")
        
        # Caso de que quiera vender algo, se añade a la lista y se pide opción nueva
        case "2":
            item = input(("Introduce que elemento quieres vender: "))
            listaVenta.append(item)
            opcion = input("Introduce tu opción:")
        
        # Caso de que quiera alquilar algo, se añade a la lista y se pide opción nueva
        case "3":
            item = input(("Introduce que es lo que quieres alquilar: "))
            listaAlquiler.append(item)
            opcion = input("Introduce tu opción:")

        # Muestra las listas por pantalla con lo que se lleva guardado
        case "4":
            print(f"LISTA COMPRA: {listaCompra}")
            print(f"LISTA VENTA: {listaVenta}")
            print(f"LISTA ALQUILER: {listaAlquiler}")
            opcion = input("Introduce tu opción:")
        
        # Cerramos el programa y nos despedimos
        case "5":
            print("Gracias por utilizar el servicio.")
            break

        # Manejo de errores en el caso de que la opción sea cualquier otra cosa. Se avisa al usuario
        case _:
            print("Opción no válida, por favor introduce una opción del 1 al 5.")
            print("MENÚ COMPRA-VENTA\n1. Comprar\n2. Vender\n3. Alquilar\n4. Mostrar listas de datos\n5. Salir")
            opcion = input("Introduce tu opción:")
```

## Ejercicio 9. Programa de Login con reintentos

Como en el resto de ejercicios, primero tenemos que establecer las variables. Necesitamos una variable para guardar la contraseña, otra para controlar el número de intentos y, por último, la variable login, que actuará como bandera para indicar al programa si el inicio de sesión ha sido correcto.

En nuestro flujo evaluamos lo siguiente: abrimos con un bucle que se repetirá hasta que el inicio de sesión sea satisfactorio, para lo cual utilizamos la variable bandera. Pedimos la contraseña y la guardamos en nuestra variable.

Después necesitamos otro bucle que se repetirá hasta 5 veces. Aquí es donde entra en juego la variable intentos, que aumentará en 1 con cada intento fallido. Por otro lado, pedimos la contraseña al usuario mediante la variable comprobarPass, que usaremos para compararla con la contraseña inicial. En caso de que coincidan, reiniciamos la variable de intentos y cambiamos login a True. Si comprobarPass es distinta, informamos al usuario y aumentamos el contador de intentos en 1. Este bucle se repetirá mientras intentos sea menor que 5.

Para reiniciar todo, añadimos una condición final que se ejecuta cuando se han agotado los intentos. En ella informamos al usuario y reiniciamos la variable de intentos. Si no hiciéramos esto, al volver a entrar en el proceso los intentos seguirían valiendo 5 y el programa se quedaría bloqueado.


```python
# Establecemos las variables que necesitamos
password = ""
intentos = 0
login = False

# Abrimos un primer bucle que evaluará si la sesión ha sido exitosa y establecemos la contraseña
while login != True:
    password = input("Introduce una contraseña:")

    # Necesitamos un segundo bucle que se repita hasta 5 veces, siempre y cuando la contraseña introducida no coincida con la establecida
    while intentos < 5:
        comprobarPass = input("Usuario 3567 introduzca su contraseña:")

        # Comprobamos la contraseña, reseteamos los valores e indicamos que la sesión ha sido exitosa por pantalla y en la variable.
        if comprobarPass == password:
            print("Contraseña correcta, has accedido al sistema.")
            intentos = 0
            login = True
            break

        
        # Si la contraseña no coincide aumentamos el valor de intentos en 1
        else:
            print("Contraseña incorrecta")
            intentos += 1
    
    # Si los reintentos llegan a 5, se reinicia el proceso y vuelve a establecer la contraseña desde el principio
    if intentos == 5:
        print("Has superado el número de intentos, vuelve a introducir la contraseña.")
        intentos = 0
        


```

## Ejercicio 10. Verificar aprobados y suspendidos con un diccionario.

Por último, vamos a trabajar con un diccionario en el que se nos dan nombres, que harán de claves, junto con su valor correspondiente, que serán las notas que han sacado. Se nos pide que recorramos el diccionario y que imprimamos si cada persona está aprobada o suspendida.

Lo primero que necesitamos es declarar la variable como diccionario; en este caso la hemos llamado notas. Después tenemos que llenarlo con las calificaciones. Lo he hecho de dos maneras distintas para comprobar que funcionaba bien: podemos añadir una clave entre corchetes y asignarle un valor, o usar el método update, que nos permite añadir un nuevo par clave‑valor al diccionario. En este caso, por ejemplo, añadimos la clave "Luis" con su correspondiente valor 4.

Una vez hecho esto, nos toca recorrer el diccionario usando un bucle. En este caso usamos for, ya que tenemos un número limitado de elementos que leer: el tamaño completo del diccionario. Si nos limitamos al enunciado, podríamos recorrer únicamente los valores y mostrar simplemente “Aprobado” o “Suspendido”. Sin embargo, quise llevarlo un poco más allá y recorrer tanto la clave como el valor para informar de quién ha aprobado o suspendido. Esto lo conseguimos con notas.items(), que nos permite obtener tanto la clave como el valor en cada iteración.

Para las condiciones necesitamos el valor, que es la nota concreta de cada alumno. Aquí también he querido refinarlo un poco para practicar, añadiendo condiciones extra relacionadas con las calificaciones posibles en un ámbito académico, es decir, notas entre 0 y 10, evitando números negativos o superiores a 10.

Por consola imprimimos con un f‑string tanto el resultado como el nombre del alumno. Por eso usamos items: para poder mostrar directamente quién ha aprobado o suspendido.


```python
# Primero declaramos el diccionario
notas = {}
# Llenamos el diccionario con nombres y notas
notas["Ana"] = 8
notas.update({"Luis": 4})
notas.update({"Marta": 6})
notas["Pedro"] = 3
notas["Daniel"] = 12

# Creamos un bucle para recorrer el diccionario y tenemos en cuenta tanto el nombre como la nota
for nombre, nota in notas.items():
    # Establecemos las condiciones y refinamos para que sean notas validas
    if nota >= 5 and nota <= 10:
        print(f"{nombre} ha aprobado")
    
    elif nota < 5 and nota >= 0:
        print(f"{nombre} ha suspendido")

    # Añadimos un manejo de notas no válidas por si ocurre cualquier error
    else:
        print(f"La nota de {nombre} no es válida")
```
