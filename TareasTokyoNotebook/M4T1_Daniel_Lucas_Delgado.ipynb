## Ejercicio 2. Lectura del fichero

Sobre el fichero anterior, obtén:
- La lectura de una sola línea del fichero.
- La lectura del fichero línea a línea.

___

````python

# TODO: La lectura de una sola línea del fichero.
def leer_linea():
    with ruta.open("r", encoding="utf-8") as f:
        print("Leyendo...")
        print(f.readline())
# TODO: La lectura del fichero línea a línea.
def leer_lineas():
    with ruta.open("r", encoding="utf-8") as f:
        print("Leyendo...")
        for linea in f:
            print("->", linea.strip())
leer_linea()
leer_lineas()
````

## Ejercicio 3. Escritura concatenando contenido

Abre el fichero en modo escritura sin sobrescribir el contenido.
- Escribe una nueva línea de texto al final del fichero.
- Comprueba que funciona correctamente cerrando y volviendo a abrir el fichero.
___

````python
def escribir_fichero():
    with ruta.open("a", encoding="utf-8") as f: # "a" Añade una línea al fichero
        print("Agregando contenido...")
        f.write("\nLínea agregada con éxito")
        # -> Al salir del bloque se cierra el archivo
escribir_fichero()
print(ruta.open("r", encoding="utf-8").read())
#-> Esta apertura está en modo lectura
````

### Función que añade línea.
Para escribir una línea sin sobreescribir todo, tenemos que abrir el archivo en modo append, es decir la letra "a".
- Cuando ejecutamos la función el fichero se cierra al salir del bloque with.
- Al abrir (en modo lectura, "r") nuevamente comprobamos que los cambios se han producido.
```python
def escribir_fichero():
    with ruta.open("a", encoding="utf-8") as f: # "a" Añade una línea al fichero
        print("Agregando contenido...")
        f.write("\nLínea agregada con éxito")
        # -> Al salir del bloque se cierra el archivo
escribir_fichero()
```

## Ejercicio 4. Escritura sobrescribiendo contenido

- Abre el fichero en modo escritura sobrescribiendo su contenido.
- Escribe una nueva línea de texto.
- Comprueba que funciona correctamente cerrando y volviendo a abrir el fichero.
___

````python
def sobreescribir_fichero():
    with ruta.open("w", encoding="utf-8") as f: # "w" sobreescribe el fichero
        print("Sobreescribiendo contenido...")
        f.write("He sobreescrito todo...")

sobreescribir_fichero()
print(ruta.open("r", encoding="utf-8").read())
````

### Sobreescribir fichero
Si cometemos el error de añadir una línea abriendo el fichero en modo write, todo lo que tuvieramos se sobreescribe con el nuevo contenido. Lo importante no es `f.write()` sino el parámetro "w" que especificamos al abrir el archivo.
```python
def sobreescribir_fichero():
    with ruta.open("w", encoding="utf-8") as f: # "w" sobreescribe el fichero
        print("Sobreescribiendo contenido...")
        f.write("He sobreescrito todo...")
```

## Ejercicio 5. Procesamiento de datos desde un fichero

- Crea un programa que lea los datos de un fichero llamado personas.txt.
- Convierte cada línea en un diccionario y añádelo a una lista llamada personas.
- Recorre la lista y muestra la información de forma amigable, por ejemplo:
    - (id=1) Carlos Pérez => 05/01/1989
- El fichero personas.txt debe contener:
    - 1;Carlos;Pérez;05/01/1989
    - 2;Manuel;Heredia;26/12/1973
    - 3;Rosa;Campos;12/06/1961
    - 4;David;García;25/07/2006
                    
- Los campos del diccionario serán: id, nombre, apellido y nacimiento.
Pista: Puede que necesites usar funciones como split() o replace() para procesar cada línea. Consulta la documentación de Python.
___

````python
def procesamiento_personas():
# Leer los datos de un fichero llamado personas.txt.
    ruta = Path("./") / "personas.txt"
    personas = []
    with ruta.open("r", encoding= "utf-8") as f:
        for linea in f:
            linea_cortada = linea.strip()
            valores = linea_cortada.split(";")
            # Convierte cada línea en un diccionario y añádelo a una lista llamada personas.
            persona = {
                # Los campos del diccionario serán: id, nombre, apellido y nacimiento.
                "id": valores[0],
                "nombre": valores[1],
                "apellido": valores[2],
                "nacimiento": valores[3]
            }
            personas.append(persona)

            #Recorre la lista y muestra la información de forma amigable, por ejemplo:
                # (id=1) Carlos Pérez => 05/01/1989
            #OUTPUT
            print(f"(id={persona["id"]}) {persona["nombre"]} {persona["apellido"]} => {persona["nacimiento"]}") #(id=1) Carlos Pérez => 05/01/1989
        print(personas)
procesamiento_personas()
````

### Línea a diccionario
Para convertir la línea en un diccionario, necesitamos crear un diccionario y asignar cada valor de la lista a su correspondiente clave. Cada posición de la lista sera un valor:
`"id": valores[0],`
Asignamos cada valor a su llave.
```python
    ruta = Path("./") / "personas.txt"
    personas = []
    with ruta.open("r", encoding= "utf-8") as f:
        for linea in f:
            linea_cortada = linea.strip()
            valores = linea_cortada.split(";")
            persona = {
                "id": valores[0],
                "nombre": valores[1],
                "apellido": valores[2],
                "nacimiento": valores[3]
            }
            print(persona) # {'id': '1', 'nombre': 'Carlos', 'apellido': 'Pérez', 'nacimiento': '05/01/1989'}
```
De esta manera conseguimos el diccionario persona.
