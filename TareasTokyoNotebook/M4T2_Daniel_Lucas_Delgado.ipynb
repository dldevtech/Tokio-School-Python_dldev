## Ejercicio 1. Lectura y almacenamiento de datos

- Observa los datos del CSV y realiza las siguientes tareas:
    - El fichero contiene 9 columnas; las 8 primeras son numéricas.
    - Crea una lista para cada una de estas columnas (mpg, cylinders, etc.).
    - No es necesario crear una lista para la columna name.
    - Recorre el CSV y almacena los datos de cada columna en su lista correspondiente.
    - Comprueba con print() que los datos se han guardado correctamente.
___

````python
import csv
from pathlib import Path
# Acceso al fichero
csv_path = Path("02_CSV_data.csv")

# Leer CSV (DictReader)
with csv_path.open("r", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f, delimiter=",")
    data = list(reader)   
# Crea una lista para cada una de estas columnas (mpg, cylinders, etc.)
# Recorre el CSV y almacena los datos de cada columna en su lista correspondiente.
mpg = [fila["mpg"] for fila in data]
cylinders = [fila["cylinders"] for fila in data]
displacement = [fila["displacement"] for fila in data]
horsepower = [fila["horsepower"] for fila in data]
weight = [fila["weight"] for fila in data]
acceleration = [fila["acceleration"] for fila in data]
year = [fila["year"] for fila in data]

# Comprueba con print() que los datos se han guardado correctamente.
print(data)
print(mpg)
print(cylinders)
print(displacement)
print(horsepower)
print(weight)
print(acceleration)
print(year)

````

Como con los archivos de la unidad anterior, accedemos a un directorio con with para evitar errores sí cometemos el error de no cerrar el stream.

- DictReader. Convierte cada fila del CSV en un diccionario usando las cabeceras del archivo como claves.
- delimiter. Importante específicar que tipo de delimitador estamos usando, en el caso de este csv es , pero podría ser ; si venímos de un archivo que proviene de excel por ejemplo.
- data. Como resultado tenemos una lista de diccionarios en los que cada una de las cabeceras del header ha sido asignado como clave y cada uno de los coches como valor. Cada diccionario representaría un coche.

## Ejercicio 3. Análisis estadístico

- Utiliza la librería statistics:
    - import statistics as stats
- Para cada lista muestra:
    - Número de observaciones.
    - Valor mínimo y máximo.
    - Media, mediana y desviación estándar.

````python
import statistics as stats
def analizar_estadisticas(nombre, lista):
    #Numero de observaciones
    #Valor mín y máx
    #Media, mediana y desviación estándar
    print(f"LISTA: {nombre}\nObservaciones: {len(lista)}\nMín:{min(lista)}\nMáx:{max(lista)}\nMedia:{stats.mean(lista)}\nMediana:{stats.median(lista)}\nDesviación estándar:{stats.stdev(lista)}\n")

analizar_estadisticas("mpg_float", mpg_float)
analizar_estadisticas("cylinders_float", cylinders_float)
analizar_estadisticas("displacement_float", displacement_float)
analizar_estadisticas("horsepower_float", horsepower_float)
analizar_estadisticas("weight_float", weight_float)
analizar_estadisticas("acceleration_float", acceleration_float)
analizar_estadisticas("year_float", year_float)
````

### Analisis estadísticos.
Se nos dice que para ciertas estadistícas tenemos que importar statistics, que es una librería que nos ofrece multitud de herramientas para analizar datos de tipo numérico.
Se nos pide:
- Número de observaciones. (len()) Es el tamaño total de elementos por lista.
- Valor mínimo y máximo. (min() y max()). Estas funciones son de python base.
- Media, mediana y desviación estándar. (stats.mean(), stats.median() y stats.stdev()). Estas funciones son de la librería statistics.
Para mostrar todo por cada una de las listas hemos creado una función a la que se le van a enviar dos parámetros, el nombre de la lista en si y la lista. Formateando todo en un f-String que nos separe todo:

## Ejercicio 4. Creación y análisis de un JSON

- Se proporciona el siguiente JSON String:

```json
{
  "jefe_proyecto": {
    "Nombre": "Juan",
    "Edad": 28,
    "Experiencia": [
      "Gestion",
      "Finanzas",
      "Bases de datos"
    ],
    "Residencia": "Madrid",
    "HorasProyecto": 3500
  },
  "empleados": {
    "Nombre": "Elena",
    "Edad": 26,
    "Experiencia": [
      "JavaScript",
      "Python"
    ],
    "Residencia": "Madrid",
    "HorasProyecto": 500
  },
  {
    "Nombre": "Luis",
    "Edad": 31,
    "Experiencia": [
      "Django",
      "Flask",
      "Pyramid"
    ],
    "Residencia": "Barcelona",
    "HorasProyecto": 1100
  }
}
```

- Crea un diccionario que reproduzca exactamente esta estructura.
- Almacénalo en una variable y muestra su tipo y contenido.
- Crea dos variables: horas_empleados y horas_jefe.
- Extrae las horas del jefe.
- Recorre los empleados y suma sus horas.
- Muestra el total de horas del proyecto.


````python
# Vete creando estructuras más pequeñas hasta que llegues a formar el mismo JSON
# string que ves arriba
empleado1 = {"Nombre": "Juan","Edad": 28,"Experiencia": ["Gestion","Finanzas","Bases de datos"],"Residencia": "Madrid","HorasProyecto": 3500}
empleado2 = {"Nombre": "Elena","Edad": 26,"Experiencia": ["JavaScript","Python"],"Residencia": "Madrid","HorasProyecto": 500}
empleado3 = {"Nombre": "Luis","Edad": 31,"Experiencia": ["Django","Flask","Pyramid"],"Residencia": "Barcelona","HorasProyecto": 1100}
# jefe_proyecto = {"jefe_proyecto": empleado1 }
# empleados = {"empleados": [empleado2, empleado3]}

# Almacena esta estructura (el diccionario) en una variable
datos_empleados = {"jefe_proyecto":empleado1,"empleados": [empleado2, empleado3]}
print(datos_empleados, type(datos_empleados))

# Crea dos variables: horas_empleados y horas_jefe
# Crear Horas para empleados
horas_empleados = 0
for empleado in datos_empleados["empleados"]:
    horas = int(empleado["HorasProyecto"])
    horas_empleados+= horas
print(f"Los empleados hicieron: {horas_empleados} horas")

# Crear horas para jefe
horas_jefe = int(datos_empleados["jefe_proyecto"]["HorasProyecto"])
print(f"El jefe hizo: {horas_jefe} horas")

# Suma todas las horas y muéstralas por pantalla
horas_proyecto = horas_empleados+horas_jefe
print(f"Las horas totales del proyecto son {horas_proyecto} horas")

````

### Estructuras pequeñas

Al imprimir su tipo y sus datos por pantalla observamos como tenemos la estructura final de diccionario.

### Creación de diccionario a través de JSON string
Un JSON String es lo mismo que lo que encontramos en un texto plano que contiene un JSON comprimido. Pero gracias al método loads, podemos transformar un JSON String en un diccionario en Pyhton.
```python
import json
datos_empleados = '''{"jefe_proyecto": {"Nombre": "Juan","Edad": 28,"Exper...}'''
datos_empleados_json = json.loads(datos_empleados)
```
Si analizamos cada una de las variables con type(variable), podemos observar que el primero se trata de un String, sin embargo, datos_empleados_json es un diccionario. Esto nos permite realizar todo tipo de acciones que hacemos en python con los diccionarios.
