## Ejercicio 1. Función para calcular el área de un rectángulo

En esta tarea solicitamos al usuario una base y una altura para calcular el área de un rectángulo. Para ello creamos una función que recibe como parámetros la base y la altura, aplica la fórmula correspondiente y devuelve el área.

Después, la función main() se encarga de pedir al usuario los valores necesarios e invocar a nuestra función.

Finalmente, llamamos a main() desde la estructura habitual de Python, siguiendo las buenas prácticas explicadas en el temario.


```python
# Lo primero que hacemos es crear la función para calcular el área
# Para ello vamos a necesitar dos argumentos
def area_rectangulo (base, altura):
    '''Calcula el área de un rectángulo'''
    area = base * altura    # Fórmula del área del rectángulo
    return area # Necesitamos retornar el valor nuevamente cuando invoquemos esta función

# Como buena práctica definimos una función main, en este caso pediremos al usuario los valores que necesitamos para calcular el área
def main():
    
    # Pedimos la base al usuario
    print("Vamos a calcular el area de un rectángulo")
    base = int(input("Introduce la base del rectángulo:"))
    
    # Pedimos la altura al usuario
    altura = int(input("Introduce la altura del rectángulo:"))
    # Creamos la variable donde vamos a guardar el área e invocamos nuestra función
    area = area_rectangulo(base, altura)
    print("El área del rectángulo es:", area)   # Mostramos por pantalla

# Finalmente tenemos que cumplir con la estructura de python e invocamos la función main desde aquí
if __name__ == "__main__":
    main()
```

## Ejercicio 2. Función para calcular el área de un círculo.
En este caso el planteamiento es prácticamente igual que en el ejercicio anterior, por lo que no es necesario extenderse demasiado. La diferencia principal es que, para obtener un resultado más preciso, en lugar de escribir manualmente el valor de pi, lo importamos desde la librería math.

Al usuario se le pide el radio del círculo y, dentro de la función, elevamos ese valor al cuadrado y lo multiplicamos por pi para obtener el área.


```python
import math

def area_circulo(radio):
    '''Calcula el área de un círculo'''
    area = (radio**2) * math.pi
    return area

def main():
    print("Vamos a calcular el area de un círculo")
    radio = int(input("Introduce el radio del círculo:"))
    area = area_circulo(radio)
    print("El área del círculo es:", area)

if __name__ == "__main__":
    main()
```

## Ejercicio 3. Función para encontrar el número mayor.

Para solucionar este ejercicio, en el propio ipynb del temario tenemos el ejemplo perfecto. En los apuntes tenemos un ejemplo en el que se piden números asi que me he fijado en esa estructura para el ejercicio. Además lo que nos piden son en parte ejercicios que ya hemos resuelto en otros temas como pedir una serie de números y almacenarlos en lista, pero en este caso estamos refactorizando el programa para dividir las responsabilidades en distintas funciones.

Creamos la función de leer números con una variable cuantos que se puede cambiar en cualquier momento, guardamos dichos números en una lista y esto lo solicitamos mediante input a través de un bucle. Al final la devolvemos con return.

Luego creamos otra función que recibe la lista de números y devuelve el número mayor utilizando la función max() de Python.
Finalmente, en la función main() llamamos a ambas funciones y mostramos el resultado al usuario.

```python
# Creamos la función que nos permite pedir los números introducidos por el usuario
def leer_numero (cuantos):
    # Los guardamos en una lista para poder exportarla con return
    numeros = []
    # Como tenemos que pedir 3 números usamos un bucle al que le vamos a pasar cuantos por si más adelante queremos pedir más numeros
    for i in range(cuantos):
        numero = int(input(f"Introduce el número {i+1}:"))  # También automatizamos qué número estamos pidiendo
        numeros.append(numero)
    
    return numeros

# Creamos la función que nos dirá cual es el número mayor de la lista
def mayor(numeros):
    n_mayor = max(numeros)
    return n_mayor

# Esta es la función principal mediante la que vamos a ejecutar el programa e invocar el resto de funciones
def main():
    print("Vamos a calcular el número máximo de tres numeros.")
    numeros = leer_numero(3)
    print(f"Tus numeros son: {numeros}")
    n_mayor = mayor(numeros)
    print(f"El número mayor es: {n_mayor}")


if __name__ == "__main__":
    main()
```

## Ejercicio 4. Cálculo del IMC.
Para calcular el IMC, el ejercicio nos proporciona la fórmula, así que lo único que necesitamos es pedir al usuario su peso y su altura en metros con decimales. Una vez obtenidos estos valores, aplicamos la fórmula del índice de masa corporal y, según el resultado, lo clasificamos en uno de los cuatro rangos establecidos. Para ello utilizamos condicionales que comprueban en qué intervalo se encuentra el valor calculado. También añadimos un else final por si los datos introducidos no son válidos.

En los condicionales hemos definido los rangos correspondientes a cada categoría. Para evitar solapamientos, usamos comparaciones encadenadas, como en el caso del rango normal: 18.5 <= imc < 25. De esta forma nos aseguramos de que cada valor cae exactamente en un único intervalo.

En la función main pedimos los valores al usuario y llamamos a la función imc, que se encarga de realizar el cálculo y mostrar el resultado. Además, mostramos el IMC con dos decimales utilizando formateo mediante f-strings, lo que permite presentar el valor de manera más clara y precisa.



```python
# Función para el imc
def imc (peso, altura):
    '''Calcula el rango según el IMC'''
    imc = peso / (altura **2)   # Fórmula del imc

    # Distintas condiciones para cada rango
    if imc < 18.5:
        print(f"Te encuentras en bajo peso, imc: {imc:.2f}")

    elif 18.5 <= imc < 25:
        print (f"Te encuentras en rango normal, imc: {imc:.2f}")

    elif 25 <= imc < 30:
        print(f"Te encuentras en sobrepeso, imc: {imc:.2f}")

    elif imc >= 30:
        print(f"Te encuentras en obesidad, imc: {imc:.2f}")

    # Condición extra en caso de comportamiento inesperado
    else:
        print("Valor no válido, recuerda que la altura es en metros con 2 decimales y el peso en kilos")

#Función main que solicita los datos por pantalla y se los manda a la función imc
def main():
    print("Vamos a calcular tu índice de masa corporal (IMC)")
    peso = float(input("Introduce tu peso en kilos:"))
    altura = float(input("Introduce tu altura en metros con 2 decimales:"))
    imc(peso, altura)


if __name__ == "__main__":
    main()
```

## Ejercicio 5. Login estructurado en funciones
En este ejercicio he reestructurado el programa original dividiéndolo en funciones pequeñas y específicas. El objetivo no ha sido optimizar el rendimiento, sino practicar la modularización y la separación de responsabilidades dentro del código.

He creado funciones independientes para cada mensaje mostrado por pantalla (contraseña incorrecta, contraseña válida y aviso de bloqueo). Aunque estos mensajes podrían haberse mantenido dentro del programa principal, he decidido externalizarlos para cumplir con el enunciado y reforzar la práctica de dividir el código en módulos.

También he separado la gestión de los intentos en dos funciones: una para incrementarlos y otra para reiniciarlos. Esto permite que el programa principal se centre únicamente en el flujo lógico, delegando las operaciones concretas en funciones auxiliares.

Por último, he externalizado la petición de contraseña, de modo que el programa principal no interactúa directamente con el usuario, sino que delega esa responsabilidad en una función dedicada.

Aunque sería posible seguir dividiendo el programa, algunas partes, como la comprobación de la contraseña dentro del bucle, resultan más difíciles de extraer sin complicar innecesariamente la estructura. Por ello, he considerado que el nivel de modularización alcanzado es adecuado y coherente con el diseño elegido.


```python
# Función para pedir la contraseñña con el mensaje
def pedir_contrasena():
    '''Pide la contraseña al usuario'''
    password = input("Introduce una contraseña:")
    return password

# ------ Funciones para los intentos------
def aumentar_intentos(intentos):
    '''Incrementa el número de intentos'''
    intentos += 1
    return intentos
def reset_intentos(intentos):
    '''Resetea el número de intentos a 0'''
    intentos = 0
    return intentos

# ------ Funciones de mensajes de sistemas ------
def mensaje_incorrecto(intentos):
    '''Imprime mensaje de contraseña incorrecta e informa del nº de intentos'''
    print(f"Contraseña incorrecta, llevas {intentos} intentos")
def mensaje_correcto():
    '''Imprime el mensaje de contraseña correcta'''
    print("Contraseña correcta, has accedido al sistema.")
def mensaje_bloqueo():
    '''Informa al usuario de que el nº de intentos se ha excedido'''
    print("Has superado el número de intentos, vuelve a introducir la contraseña.")

# Función principal que orquesta todo
def main ():  
    # Establecemos las variables que necesitamos
    password = ""
    intentos = 0
    login = False

    # Abrimos un primer bucle que evaluará si la sesión ha sido exitosa y establecemos la contraseña
    while login != True:
        password = pedir_contrasena()

        # Necesitamos un segundo bucle que se repita hasta 5 veces, siempre y cuando la contraseña introducida no coincida con la establecida
        while intentos < 5:
            comprobarPass = input("Usuario 3567 introduzca su contraseña:")

            # Comprobamos la contraseña, reseteamos los valores e indicamos que la sesión ha sido exitosa por pantalla y en la variable.
            if comprobarPass == password:
                mensaje_correcto()
                intentos = reset_intentos(intentos)
                login = True
                break

            
            # Si la contraseña no coincide aumentamos el valor de intentos en 1
            else:
                intentos = aumentar_intentos(intentos)
                mensaje_incorrecto(intentos)
        
        # Si los reintentos llegan a 5, se reinicia el proceso y vuelve a establecer la contraseña desde el principio
        if intentos == 5:
            mensaje_bloqueo()
            intentos = reset_intentos(intentos)

if __name__ == "__main__":
    main()
        
```

## 6. Función intermedio
En este ejercicio necesitamos calcular el valor intermedio entre dos números. Primero he creado la función intermedio, que recibe dos números y devuelve su valor medio. La operación consiste simplemente en sumar ambos valores y dividir el resultado entre dos.

Aunque el enunciado no lo exige, he decidido externalizar también la petición de números en la función pedir_numeros, para mantener el programa principal más limpio y modular. Esta función se encarga de solicitar al usuario los dos valores necesarios y devolverlos.

Por último, en la función main he reunido el flujo principal del programa: se muestra un mensaje inicial, se obtienen los números llamando a pedir_numeros, se calcula el valor intermedio mediante la función intermedio y se muestra el resultado al usuario.

Con esta estructura, cada función tiene una responsabilidad concreta y el programa queda más organizado y fácil de entender.


```python
# Función que calcular el valor intermedio entre dos números
def intermedio(n1, n2):
    '''Calcula el número intermedio'''
    intermedio = (n1+n2)/2
    return intermedio

# Función que solicita al usuario los números sobre los que hacemos la operación
def pedir_numeros():
    '''Pide dos números al usuario'''
    n1 = int(input("Introduce el primer número:"))
    n2 = int(input("Introduce el segundo número:"))
    return n1, n2

# Función principal que organiza el flujo principal
def main():
    print("Vamos a calcular el número intermedio entre dos números.")   # Mensaje inicial informativo
    n1, n2 = pedir_numeros()
    n_intermedio = intermedio(n1, n2)
    print(f"El número intermedio entre {n1} y {n2} es: {n_intermedio}") # Mensaje con el resultado final

if __name__ == "__main__":
    main()
```

En este ejercicio he decidido añadir un nivel extra de control respecto a lo que pedía el enunciado, y explico a continuación el motivo. La función principal, recortar(), recibe un número y dos límites (inferior y superior). Su objetivo es devolver el número ajustado para que quede dentro del rango definido por esos límites. Para ello se evalúan varios casos mediante condicionales:

    - Si el número es menor que el límite inferior, se devuelve el límite inferior.

    - Si el número es mayor que el límite superior, se devuelve el límite superior.

    - Si el número está dentro del rango, se devuelve sin modificar.

    - Si los límites están mal definidos (por ejemplo, si el límite inferior es mayor que el superior), la función no debería continuar porque el resultado sería incoherente.

Para gestionar este último caso he añadido una función adicional llamada comparar_limites(). Esta función comprueba si el límite inferior es realmente menor o igual que el límite superior. De esta forma, en el flujo principal del programa puedo verificar primero que los límites son válidos antes de llamar a recortar(). Si el usuario introduce los límites al revés, el programa muestra un mensaje de error en lugar de devolver un valor incorrecto.

Sin esta comprobación previa, el programa podría devolver resultados erróneos cuando los límites se introducen en orden incorrecto, ya que la función recortar() asumiría que los límites son coherentes.

```python
# Función recortar en el que se establecen tres condiciones distintas
def recortar(num_recorte, limite_inferior, limite_superior):
    '''Recorta un número en base a dos límites'''
    if num_recorte < limite_inferior:
        return limite_inferior
    
    elif num_recorte > limite_superior:
        return limite_superior
    
    else:
        return num_recorte

# Función preventiva para evitar que el usuario no coloque los límites de forma incorrecta    
def comparar_limites(limite_inferior, limite_superior):
    '''Comprueba que los límites están en el orden correcto'''
    if limite_inferior <= limite_superior:
        return True
    else:
        return False

# Función principal del programa
def main():
    # Instrucciones y almacenamiento de variables
    print("Vamos a recortar un número para que se quede dentro de un rango.")
    num_recorte = int(input("Introduce el número a recortar:"))
    limite_inferior = int(input("Introduce el límite inferior:"))
    limite_superior = int(input("Introduce el número superior:"))
    # Comprobación de límites, si todo es correcto 
    if comparar_limites(limite_inferior, limite_superior):
        recortado = recortar(num_recorte, limite_inferior, limite_superior)
        print(f"El número recortado es: {recortado}")
    
    # En caso contrario imprimimos el resultado de lo que se ha introducido
    else:
        print("Ha habido un error al introducir los datos, prueba de nuevo.")
        print(f"Numero recortado: {num_recorte}\nLimite inferior: {limite_inferior}\nLimite superior: {limite_superior}")
    
    
if __name__ == "__main__":
    main()
```

Este ejercicio es más sencillo porque la función no solicita datos al usuario, lo que elimina posibles errores de entrada. Como ya disponemos de una lista de números, únicamente necesitamos construir la función que los separe.

La función separar() recibe una lista de enteros y debe devolver dos listas distintas: una con los números pares y otra con los impares. Para ello, recorremos la lista con un bucle for y evaluamos cada número. Si el número es par (su resto al dividir entre 2 es cero), lo añadimos a la lista de pares; en caso contrario, lo añadimos a la lista de impares.

Al finalizar el recorrido, devolvemos ambas listas, primero la de pares y luego la de impares. Es importante mantener este orden para que, al invocar la función en el main, las variables reciban correctamente cada lista.

Una vez obtenidas las dos listas, las ordenamos con .sort() y las mostramos por pantalla. De esta forma, el resultado final presenta los números pares e impares ordenados de menor a mayor.

```python
def separar(numeros):
    '''Separa números entre pares e impares'''
    numeros_pares = []
    numeros_impares = []
    for numero in numeros:
        if numero % 2 == 0:
            numeros_pares.append(numero)
        else:
            numeros_impares.append(numero)
    
    return numeros_pares, numeros_impares


def main():
    # La lista de números que nos dan de ejemplo
    numeros = [-12, 84, 13, 20, -33, 101, 9]
    # Invocamos a la función y ordenamos los números
    numeros_pares, numeros_impares = separar(numeros)
    # Ordenamos las listas
    numeros_pares.sort()
    numeros_impares.sort()
    # Mostramos el resultado por pantalla
    print(f"Los números pares ordenados son: {numeros_pares}")
    print(f"Los números impares ordenados son: {numeros_impares}")

if __name__ == "__main__":
    main()
```
