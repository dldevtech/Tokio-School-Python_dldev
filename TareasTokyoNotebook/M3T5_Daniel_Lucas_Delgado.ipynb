## Ejercicio 1. Transformaciones básicas con map y lambda

Creación:
    - Parte de una lista de números enteros, por ejemplo: [1, 2, 3, 4, 5].
    - Utiliza map y lambda para generar:
        - Una lista con los cuadrados de los números.
        - Una lista con los dobles.
        - Una lista de cadenas con el formato "N=<valor>".

Experimentación:
    - Muestra las tres listas resultantes.
    - Comprueba que las longitudes de las tres listas coinciden con la de entrada.
    - (Opcional) Aplica una segunda transformación con map a la lista de cadenas para añadir un sufijo, por ejemplo "(ok)".

`````python
listaEnteros = [1, 2, 3, 4, 5]
# Lambda Function para comprobar longitudes entre listas
comprobacion = lambda lista1, lista2: f"Longitud entre listas es igual: {len(lista1) == len(lista2)}\n"

# Map y Lambda para Cuadrados
listaCuadrados = list(map(lambda num: num * num, listaEnteros))
print("listaCuadrados:", listaCuadrados)
print(comprobacion(listaEnteros, listaCuadrados))

# Map y Lambda para Dobles
listaDobles = list(map(lambda num: num*2, listaEnteros))
print("listaDobles:", listaDobles)
print(comprobacion(listaEnteros, listaDobles))

# Map y Lambda para formato "N=<valor>"
listaCadenas = list(map(lambda num: f'N = {num}', listaEnteros))
print(listaCadenas)
print(comprobacion(listaEnteros, listaCadenas))

# Map opcional de doble trasnformacion de la listaCadenas con sufijo ok
listaCadenas2 = list(map(lambda string: f"{string}(ok)", listaCadenas))
print(listaCadenas2)
`````

### Map y Lambda
En este caso queremos mostrar una lista como resultado al crear cada una de las transformaciones que nos piden. Para ello:
    -Usamos list(), lo que nos permite convertir a lista lo que arroje la lambda function.
