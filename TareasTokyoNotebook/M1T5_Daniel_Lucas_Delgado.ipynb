## 1.- Listas y Tuplas
Las listas son como cajones en los que podemos guardar distintos tipos de datos, podemos modificar ítems dentro de la lista y realizar lo que se llaman operaciones CRUD con ellas. Por otro lado las tuplas son cajones pero inmutables, se suelen utilizar sobre todo para manejar datos que sabemos que no van a variar.

````python
# Primero guardaremos nuestro equipo Pokemon en una lista

equipoPokemon = ["Charizard", "Umbreon", "Lucario", "Sylveon", "Jolteon"]

# Y a continuación una Tupla, por ejemplo las estadísticas base que tienen cada uno de nuestros pokemon, ya que no cambian nunca

statsPokemon = ("Ps", "Atq", "Def", "AtqEsp", "DefEsp", "Vel")

# Mostramos por pantalla
print("Mi equipo Pokemon es:", equipoPokemon)

print("Las estadísticas base de un Pokemon son siempre:", statsPokemon)

# Para acceder al segundo elemento de la lista, y también de una tupla, tenemos que indicar el índice, entendiendo que este comienza por 0

print(f"Mi segundo Pokemon será:", equipoPokemon[1])

# En el caso de las estadísticas sabemos que el -1 es la última posición de este tipo de cajones por lo que la penúltima sera -2
print(f"La penúltima estadística de un Pokemon es:",statsPokemon[-2])

# Modificación de elementos
equipoPokemon[0]= "Typhlosion"
print(f"He cambiado a Charizard, ahora mi equipo es:", equipoPokemon)

# Nos lanza un error al modificar la tupla, por ello lo comentaré, pero se ha realizado una captura
#statsPokemon [0] = "Vida"
#print(statsPokemon)

# Para saber la longitud podemos usar len y nos arroja el número total de elementos
print("El tamaño del equipo Pokemon es: ", len(equipoPokemon))
print("El número de stats que tiene un Pokemon es: ", len(statsPokemon))

# Para la búsqueda dentro de una lista o una tupla se hace de la misma manera
print("¿Está Charizard en mi equipo actualmente? ","Charizard" in equipoPokemon)

print("¿Los Pokemon tienen Ps? ", "Ps" in statsPokemon)

# Vamos a añadir un Pokemon a nuestro equipo en este caso un Mewtwo por ejemplo, esto lo hacemos con el método append
equipoPokemon.append("Mewtwo")
print("Mi equipo Pokemon ahora es:", equipoPokemon)

# En este caso vamos a añadir una estadística a la tupla aunque nos va a dar error así que lo mantengo comentado
#statsPokemon.append("Suerte")

# Eliminación de elementos, aunque en el caso de las tuplas no será posible mostraremos los errores en capturas y comentaremos el código
# Hay maneras de eliminar el contenido específico de la lista, como indicando el índice o por el valor en concreto, pero se nos pide que eliminemos la lista entera

equipoPokemon.clear()
print(equipoPokemon)

#statsPokemon.clear()
#print(statsPokemon)
````

## 2.- Sets y Diccionarios

Seguiremos con la misma temática de Pokemon ya que puede ser interesante de cara a mostrar los distintos ejemplos que nos plantean y para salirnos de la convención de persona o coche.



````python
# Primero creamos un conjunto, que podría ser los Pokemon que tenemos avistados en la Pokedex por ejemplo por su número
pokemonPokedex = {4, 5, 6, 23, 37, 150,}

# Además de ello un diccionario con datos que puede tener un Pokemon
datosPokemon = {"Nombre": "Charizard", "Tipo": "Fuego/Volador", "Ps": 78, "Atq": 84, "Def": 78, "AtqEsp": 109, "DefEsp": 85, "Vel": 100}

# Mostramos por pantalla
print(pokemonPokedex)
print(datosPokemon)

# Mostramos el segundo elemento del set, es decir la posición 1 y el valor del primer clave/valor del diccionario
#print(pokemonPokedex[1])
print(datosPokemon["Nombre"])

# Modificación de valores
datosPokemon["Ps"] = 110
print(datosPokemon)

#pokemonPokedex[0] = 1

# Mostrar el tamaño específico del conjunto y el diccionario
print(len(pokemonPokedex))
print(len(datosPokemon))

# Búsqueda de un elemento en conjuntos y diccionarios
print("¿Está el Pokemon número 150 en la Pokedex?", 150 in pokemonPokedex)
print("¿Tiene el Pokemon Charizard la clave 'Naturaleza'?", "Naturaleza" in datosPokemon)

# A continuación añadiremos elementos a ambas variables
pokemonPokedex.add(151)
datosPokemon["Naturaleza"] = "Firme"
print(pokemonPokedex)
print(datosPokemon)

# Por último borraremos todo el contenido como se sugiere en el ejercicio y lo mostraremos
# Para el caso de los conjuntos lo hacemos llamando a clear
pokemonPokedex.clear()
print(pokemonPokedex)

# Para el caso de los diccionarios llamamos al mismo método
datosPokemon.clear()
print(datosPokemon)
````

## 3.- Suma de Elementos de una Lista
En este ejercicio nos piden que le pidamos a un usuario tres números flotantes para mostrar el resultado por pantalla. También se especifíca que no hace falta usar buccles ya que no se han usado aún.



````python
numeros = []
numeros.append(float(input("Introduce el primer número flotante, recuerda poner el . después de las unidades: ")))
numeros.append(float(input("Introduce el segundo número flotante, recuerda poner el . después de las unidades: ")))
numeros.append(float(input("Introduce el tercer número flotante, recuerda poner el . después de las unidades: ")))
print("Los números introducidos son:", numeros)

sumatorio = sum(numeros)
print("El resultado total es: ", sumatorio)
````

Al empezar, me encontré con varios errores que me ayudaron a entender mejor cómo funciona input(), las listas y el método append().

*Primer intento:* sobrescribiendo la lista sin querer.

Inicialmente convertía cada input() a float, pero lo asignaba directamente a la variable numeros. Esto hacía que la lista se perdiera y que numeros pasara a ser un único número. Cuando intenté usar sum(numeros), Python me devolvió un error porque un float no es iterable. En ese momento entendí que si quiero sumar varios valores, necesito mantenerlos dentro de una lista, no reemplazarla.

*Segundo intento:* intentando asignar al método append.

Después intenté usar append,de la misma manera que hice anteriormente con "Mewtwo" pero cometí el error de escribirlo como si fuera una variable ya que me confundí debido a la petición del input:

```python
numeros.append = float(input(...))
```

Esto provocó otro error, porque append no se puede sobrescribir: es un método de solo lectura. Aquí comprendí que append no se asigna, sino que se usa con paréntesis para meter un valor dentro de la lista.

Al final la clave estaba en combinar correctamente las dos cosas:
- convertir el valor introducido por el usuario a float,
- y meter ese valor dentro de la lista usando append().
De esta forma, la lista mantiene los tres números flotantes y sum() puede trabajar con ellos sin problemas.

## 5.-Manipulación de lista de la compra.

A continuación vamos a trabajar con lista de strings y a ejecutar distintas acciones como slicing, modificación y selección por índice.

````python
listaCompra =["Poción", "Revivir", "Restaura Todo", "PokeBall", "SuperBall", "Antídoto"]

# Para mostrar el segundo y el último emplearemos fString e indicaremos los índices de la lista
print(f"El segundo producto de la lista es {listaCompra[1]}, mientras que el último sería {listaCompra[-1]}")

# Ahora modificamos el tercer elemento, en este caso el índice 2
listaCompra[2] = "Caramelo Raro"
print(f"La lista de la compra ahora es: {listaCompra}")

# Para intercambiar el primer y el último lugar se me ocurre usar reverse() pero eso modificaría toda la lista
# En vez de eso me aprovechare de las bondades de la sintaxis de Python

listaCompra[0], listaCompra[-1] = listaCompra[-1], listaCompra[0]
print(f"Ahora mi lista de la compra está modificada solamente en el primer y el último lugar: {listaCompra}")

print("Para extraer los tres primeros elementos lo indicamos con la técnica slicing [0:3] sabiendo que el índice 3 no aparece", listaCompra[0:3])

print("Para extraer los tres últimos lo hacemos de forma parecida pero usando números negativos [-3:] si dejamos en blanco llega al final", listaCompra[-3:])

# Para recorrer desde los pares tenemos que empezar desde la segunda posición que es 1, por ello [1::2] avanza desde el segundo y cada dos posiciones
print("En el caso de mostrar elementos pares podemos establecer un inicio, un fin y cada cuantos avanzamos [1::2] dejamos el fin indeterminado para llegar al final", listaCompra[1::2])
````

En el último ejercicio se nos pedía extraer los elementos que están en posiciones pares de la lista. Yo lo he interpretado como posiciones pares en el sentido humano (2ª, 4ª, 6ª…), lo que corresponde a los índices 1, 3, 5… en Python. Por eso he usado el slicing [1::2], que empieza en el índice 1 y avanza de dos en dos.
Si en cambio habláramos de posiciones pares según los índices de Python (0, 2, 4…), entonces el slicing correcto sería [0::2], empezando en el índice 0 y avanzando también de dos en dos.

## 6.- Formateo de cadenas invertidas

En este ejercicio invertimos la cadena original usando slicing con , tal como sugiere el enunciado. Una vez invertida, la cadena queda en el orden correcto para extraer primero el año y después el nombre completo. Para ello usamos slicing: los cuatro primeros caracteres  corresponden al año, y a partir del índice 6  obtenemos el nombre sin la coma ni el espacio. Finalmente mostramos el resultado con una f-string.


````python
cadena = "zereP solraC, 5891"

# Para invertir la cadena el propio ejercicio nos sugiere que usemos slicing con [::-1]
cadenaInvertida = cadena[::-1]
print (cadenaInvertida)

# Ahora deberiamos separar el año del nombre completo para cumplir con lo que nos piden en el ejercicio:
# Empleado Nombre Apellido nacido en el año Año
# Para hacer eso podemos cortar con el slicing como hemos parendido

fecha = cadenaInvertida[0:4]
nombreCompleto = cadenaInvertida[6:]
print(fecha)
print(nombreCompleto)
print(f"Empleado {nombreCompleto} nacido en el año {fecha}")
````

## 7.- Operaciones con conjuntos
En este ejercicio realizamos operaciones muy sencillas sobre los conjuntos, como borrar o añadir elementos.

````python
# Primero creamos los dos conjuntos que nos propone el ejercicio

mamiferos = {"Gato", "León","Caballo", "Ballena", "Tortuga"}
peces = {"Ballena", "Sardina"}

# Ahora borramos Tortuga con la función remove
mamiferos.remove("Tortuga")

# Añadimos Pez Payaso con add
peces.add("Pez Payaso")

# Imprimimos el resultado de ambos conjuntos
print(mamiferos)
print(peces)

# Para mostrar los elementos que estan en ambos conjuntos usamos & que nos da la intersección de ambas variables
print(mamiferos&peces)
````

Lo que hemos hecho en esta tarea para empezar es crear dos conjuntos distintos. Para crear un conjunto establecemos el nombre de la variable, en nuestro caso mamiferos y peces, después igualamos con un {} y dentro añadimos los elementos que queramos. Lo importante a tener en cuenta es que no podemos tener elementos duplicados y que podemos realizar acciones como las que se nos han pedido. 
Al contrario que en las tuplas este tipo de variable no es inmutable, pero tampoco podemos acceder a posiciones concretas de un índice ya que no siempre muestra el mismo orden.
Lo que sí podemos hacer es:
- Evitar duplicidad.
- Añadir o eliminar un elemento concreto.
- Realizar operaciones de unión, intersección o diferencia entre conjuntos.

En el ejercicio se nos proponía que realizaramos los elementos comunes de peces y mamíferos, por eso aparece la ballena con el &.

## 9.- Depurando una lista de duplicados
En este ejercicio vamos a depurar listas para que no haya duplicados, algo bastante importante de cara al manejo de datos en bases de datos. Seguiremos con el ejemplo de Pokemon como en las otras partes libres.

````python
# Tenemos una lista en el que vamos a guardas los encuentros con los Pokemon cuando navegamos por el mapa
encuentrosPokemon = ["Ratata", "Pidgey", "Pidgey", "Ratata", "Spearow", "Caterpie", "Caterpie"]

# Para eliminar los duplicados pasamos la lista a conjunto pero se nos pide que lo guardemos nuevamente en una "nueva" lista por eso anidamos un casting dentro de otro
pokemonAvistados = list(set(encuentrosPokemon))
print("Los Pokemon que he avistado son: ", pokemonAvistados)
````

## 10.- Gestión de tareas con un diccionario
Vamos a realizar los distintos puntos en esta tarea, que consiste en crear un diccionario de tareas que esten en tres claves distintas: pendientes, completadas o en progreso


````python
# Primero creamos el diccionario que tendrá tres claves distintas: pendiente, enProgreso y completada. Y lo separamos en líneas distintas para que sea más visual
tareas= {"pendiente": [],
         "enProgreso":[],
         "completada": []
         }

# Ahora se le pide al usuario que introduzca los datos por consola, esto lo hacíamos con input y con add
nombreTarea = input("Introduce el nombre de la tarea: ")
estado = input("Introduce el estado de la tarea, puede ser completada, pendiente o progreso: ")

# Ahora añadimos los elementos al diccionario
tareas[estado].append(nombreTarea)
# Mostramos el diccionario actualizado 
print("Diccionario actualizado: ", tareas)
````

Como hemos explicado anteriormente lo que hacemos es generar un diccionario que va a conetener listas de tareas en base a su estado, por eso tenemos que distinguir en clave: "estado de la tarea" y la lista vacía para las descripciones.
Después al realizar el input lo guardamos en variables independientes para finalmente añadirlo a la lista dependiendo de su estado. Por este motivo primero ponemos tareas[estado] y hacemos .append con la descripcion de la tarea para añadirla a la lista.
